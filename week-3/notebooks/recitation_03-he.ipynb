<h2 id="תזכורת-קצרה">תזכורת קצרה</h2>
<p dir="rtl" style="text-align:right">אנו מאתחלים מצב, בודקים תנאי <code dir="ltr">while</code>, מעדכנים את המצב ומנמקים עצירה. וקטורים הם בעלי אינדוקס המתחיל ב-1; נשתמש ב-<code dir="ltr">end</code>, <code dir="ltr">length</code>, <code dir="ltr">eachindex</code> ו-<code dir="ltr">push!</code>. עבור <code dir="ltr">mean</code> ו-<code dir="ltr">std</code> נטען את <code dir="ltr">Statistics</code>.</p>
<h2 id="בעיית-התרגול-המונחית-עד-20-דקות">בעיית התרגול המונחית (עד 20 דקות)</h2>
<h4 id="קירור-נדגם">קירור נדגם</h4>
<p dir="rtl" style="text-align:right">נמודל גוף שהפרש הטמפרטורה שלו מסביבה קבועה מוכפל ב-0.9 בכל דקה. באופן כללי <span dir="ltr" class="math inline">\(T_{n+1}=T_{\rm room}+0.9(T_n-T_{\rm room})\)</span>. בדוגמה נבחר <span dir="ltr" class="math inline">\(T_{\rm room}=0^\circ\mathrm C\)</span>, ולכן <span dir="ltr" class="math inline">\(T_{n+1}=0.9T_n\)</span>. אלה דגימות מדויקות של קירור ניוטון בקצב <span dir="ltr" class="math inline">\(-\log(0.9)\)</span> לדקה. המודל מניח טמפרטורה אחידה בגוף ואיבוד חום לינארי בקירוב; הכפלת טמפרטורת צלזיוס שרירותית ב-0.9 ללא קביעת הסביבה מטעה פיסיקלית. נחפש את הדגימה הראשונה על הסף או מתחתיו; שוויון נחשב הגעה.</p>
</div>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>לפני ההרצה:</strong> נחזה את הערך, הטקסט או הגרף המתוארים לעיל. זו הדוגמה המתאימה מהפרק הכתוב.</p>
</div>

```julia
let temperature = 100.0   # C
threshold = 81.0          # C
minute = 0
while temperature > threshold
    temperature = 0.9 * temperature
    minute = minute + 1
end
println((minute, temperature))
end
```

<h2 id="מעקב-ועבודה-נתמכת">מעקב ועבודה נתמכת</h2>
<p dir="rtl" style="text-align:right">נעקוב אחרי <span dir="ltr" class="math inline">\((0,100)\)</span>, אחר כך <span dir="ltr" class="math inline">\((1,90)\)</span> ולבסוף <span dir="ltr" class="math inline">\((2,81)\)</span>. לאחר מכן נתחיל את תרגיל הבית: נשמור מסלול ממרפסת ונשווה שני צעדי זמן. התרגיל הוא קטע קוד כי הגדרת פונקציות מתחילה בשבוע 5.</p>
<h2 id="תרגול-זמן-הדגימה-הראשון-של-סף-קירור">תרגול: זמן הדגימה הראשון של סף קירור</h2>
<p dir="rtl" style="text-align:right">נשתמש במודל הקירור של ניוטון, <span dir="ltr" class="math inline">\(\dot T=-k(T-T_{\rm room})\)</span>, עבור גוף שהטמפרטורה בתוכו אחידה בקירוב וסביבתו קבועה. הקצב <span dir="ltr" class="math inline">\(k&gt;0\)</span> ביחידות שנייה הפוכה; הפרשי טמפרטורה נמדדים בקלווין או במעלות צלזיוס. נניח שקצב איבוד החום יחסי בקירוב להפרש הטמפרטורות, ולא נטען שהמודל מתאר התאדות או השפעות קרינה גדולות.</p>
<p dir="rtl" style="text-align:right">מחברת התרגול מספקת את העדכון הפיסיקלי. נזהה את האתחול, התנאי, העדכון והמונה, ונשתמש בלולאת <code dir="ltr">while</code> למציאת הדגימה הראשונה על הסף או מתחתיו. שוויון נחשב הגעה לסף. נחזה מה קורה אם הדגימה ההתחלתית כבר מתאימה, ומדוע קירור מונוטוני מלמעלה אינו מגיע לסף הנמוך מטמפרטורת החדר. לאחר דיון של עשרים דקות נתחיל את מטלת המרפסת.</p>
<h2 id="מטלת-החובה-זריקה-ממרפסת">מטלת החובה: זריקה ממרפסת</h2>
<p dir="rtl" style="text-align:right">נרחיב את דוגמת הקליע הנדגם לגובה, מהירות, זווית וצעד זמן חיובי שסופקו כקלט. נשמור על <span dir="ltr" class="math inline">\(+y\)</span> כלפי מעלה ונקדם את המצב הישן בשיטת אוילר המפורשת. נשמור היסטוריות תואמות של זמן, מיקום ושני רכיבי המהירות, לרבות המצב ההתחלתי והדגימה הראשונה שבה <span dir="ltr" class="math inline">\(y\le0\)</span>.</p>
<p dir="rtl" style="text-align:right">נעבוד בארבעה שלבים. תחילה נשרטט מסלול צפוי, נקבע יחידות וסימנים ונחשב זמן נחיתה אנליטי להשוואה. לאחר מכן נבנה את הווקטורים ואת הלולאה ונבדוק ידנית את העדכון הראשון. נוודא שכל הווקטורים באותו אורך, שהזמן מתקדם בצעד שנקבע, שהמהירות האופקית קבועה ושהמהירות האנכית קטנה ב-<span dir="ltr" class="math inline">\(g\Delta t\)</span>. לבסוף נחזור עם חצי צעד ונציג בטבלה את זמני וטווחי הנחיתה הנדגמים לצד הערכים האנליטיים.</p>
<p dir="rtl" style="text-align:right">בדוח נבחין בין דגימת חציית קרקע לבין פגיעה מדויקת או פגיעה שחושבה באינטרפולציה. לא נאפס את הגובה השלילי האחרון ולא נחליף בשקט את התוצאה הנומרית באנליטית. נפעל לפי מוסכמת המצב ההתחלתי בדף המטלה, לרבות שיגור כלפי מעלה מהקרקע; התנאי <code dir="ltr">while y[end] &gt; 0</code> לבדו אינו מטפל בכל תנאי התחלה. דף המטלה מפרט את קטע החישוב, את הבדיקות הגלויות ואת תוצרי המסלול ובדיקת צעד הזמן. הגדרת פונקציות נשארת חומר של שבוע 5.</p>
<h2 id="בונוס-לבחירה-הילוך-מקרי-שניתן-לשחזר">בונוס לבחירה: הילוך מקרי שניתן לשחזר</h2>
<p dir="rtl" style="text-align:right">הבונוס הוא רשות, אינו תנאי להמשך ואינו נבחן. הוא מזכה ב<strong>שליש נקודה בציון הסופי</strong>.</p>
<p dir="rtl" style="text-align:right">נקרא את המבוא המסופק למחוללים פסאודו-אקראיים ונבנה הילוך דו-ממדי על סריג. כל צעד הוא יחידה אחת בדיוק צפונה, דרומה, מזרחה או מערבה. עבור <span dir="ltr" class="math inline">\(n\)</span> צעדים נשמור <span dir="ltr" class="math inline">\(n+1\)</span> איברים בכל וקטור, משום שגם הראשית היא דגימה. המחולל המסופק מאפשר לשחזר את המסלול לצורך בדיקה; הוא אינו הופך את ההילוך לחוק פיסיקלי דטרמיניסטי.</p>
<p dir="rtl" style="text-align:right">תחילה נעקוב ידנית אחר כמה כיוונים נתונים. לאחר מכן נכתוב את הלולאה והענפים, נריץ את הדוגמאות בעלות הזרע הקבוע ונבדוק ראשית, אורכים, קואורדינטות שלמות והעתקי יחידה בכיווני הצירים. נשתמש במחולל שהועבר לקטע ולא נאפס זרע גלובלי בכל צעד. האנלוגיה הפיסיקלית היא רצף אידיאלי של העתקים בלתי תלויים; לתנועה מולקולרית אמיתית יש סקלות זמן ואורך ומתאמים שהמודל הפשוט משמיט.</p>
<h2 id="להעמקה-מהילוך-יחיד-לדיפוזיה">להעמקה: מהילוך יחיד לדיפוזיה</h2>
<p dir="rtl" style="text-align:right">מסלול יחיד יכול לחזור לעבר הראשית. דיפוזיה היא טענה על צבר: עבור צעדי יחידה בלתי תלויים וחסרי הטיה, ממוצע ריבוע ההעתק מקיים <span dir="ltr" class="math inline">\(\langle r_n^2\rangle=n\)</span> ביחידות הסריג, משום שאיברי המכפלה מתאפסים בממוצע. אם משך צעד הוא <span dir="ltr" class="math inline">\(\tau\)</span> ואורכו <span dir="ltr" class="math inline">\(\ell\)</span>, מקדם הדיפוזיה הדו-ממדי המתאים הוא <span dir="ltr" class="math inline">\(D=\ell^2/(4\tau)\)</span>.</p>
<p dir="rtl" style="text-align:right">בעזרת מתכון מסופק נשווה צבר קטן לצבר גדול יותר, נשרטט ממוצע ריבוע העתק מול מספר הצעד ונבחן תנודות סביב הישר הצפוי. נסביר מדוע ריבוע ההעתק הממוצע הוא גודל אחר, העשוי להישאר קרוב לאפס. נציין לצד הגרף את הגדרת המחולל ואת גודל הצבר. החקירה היא רשות ואינה נבחנת; הכלים המסופקים לעבודה עם צבר אינם תנאי לחומר החובה.</p>
