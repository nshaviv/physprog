<h2 id="תזכורת-קצרה">תזכורת קצרה</h2>
<p dir="rtl" style="text-align:right">אנו מאתחלים מצב, בודקים תנאי <code dir="ltr">while</code>, מעדכנים את המצב ומנמקים עצירה. וקטורים הם בעלי אינדוקס המתחיל ב-1; נשתמש ב-<code dir="ltr">end</code>, <code dir="ltr">length</code>, <code dir="ltr">eachindex</code> ו-<code dir="ltr">push!</code>. עבור <code dir="ltr">mean</code> ו-<code dir="ltr">std</code> נטען את <code dir="ltr">Statistics</code>.</p>
<h2 id="בעיית-התרגול-המונחית-עד-20-דקות">בעיית התרגול המונחית (עד 20 דקות)</h2>
<h4 id="קירור-נדגם">קירור נדגם</h4>
<p dir="rtl" style="text-align:right">נמודל גוף שהפרש הטמפרטורה שלו מסביבה קבועה מוכפל ב-0.9 בכל דקה. באופן כללי <span dir="ltr" class="math inline">\(T_{n+1}=T_{\rm room}+0.9(T_n-T_{\rm room})\)</span>. בדוגמה נבחר <span dir="ltr" class="math inline">\(T_{\rm room}=0^\circ\mathrm C\)</span>, ולכן <span dir="ltr" class="math inline">\(T_{n+1}=0.9T_n\)</span>. אלה דגימות מדויקות של קירור ניוטון בקצב <span dir="ltr" class="math inline">\(-\log(0.9)\)</span> לדקה. המודל מניח טמפרטורה אחידה בגוף ואיבוד חום לינארי בקירוב; הכפלת טמפרטורת צלזיוס שרירותית ב-0.9 ללא קביעת הסביבה מטעה פיסיקלית. נחפש את הדגימה הראשונה על הסף או מתחתיו; שוויון נחשב הגעה.</p>
</div>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>לפני ההרצה:</strong> נחזה את הערך, הטקסט או הגרף המתוארים לעיל. זו הדוגמה המתאימה מהפרק הכתוב.</p>
</div>

```julia
begin temperature = 100.0   # C
threshold = 81.0          # C
minute = 0
while temperature > threshold
    temperature = 0.9 * temperature
    minute = minute + 1
end
println((minute, temperature))
end
```

<h2 id="מעקב-ועבודה-נתמכת">מעקב ועבודה נתמכת</h2>
<p dir="rtl" style="text-align:right">נעקוב אחרי <span dir="ltr" class="math inline">\((0,100)\)</span>, אחר כך <span dir="ltr" class="math inline">\((1,90)\)</span> ולבסוף <span dir="ltr" class="math inline">\((2,81)\)</span>. לאחר מכן נתחיל את תרגיל הבית: התוכנית קוראת חמישה קלטים, מחקה זריקת פטיש עם גרר ריבועי עד לדגימה המוצעת הראשונה מתחת לקרקע או עד זמן <span dir="ltr" class="math inline">\(T\)</span>, מדפיסה את מרחק הנחיתה ושומרת את איור המסלול. הלולאה משתמשת באותם רעיונות של גידול מערך ועצירה כמו לולאת הקירור שלעיל; הגדרת פונקציות מתחילה בשבוע 5, כך שהתוכנית אינה משתמשת בהן.</p>
<h2 id="תרגול-זמן-קירור-נדגם-ראשון">תרגול: זמן קירור נדגם ראשון</h2>
<p dir="rtl" style="text-align:right">אנו משתמשים במודל הקירור של ניוטון, <span dir="ltr" class="math inline">\(\dot T=-k(T-T_{\rm room})\)</span>, עבור גוף עם טמפרטורה פנימית אחידה בקירוב בסביבה קבועה. הקצב <span dir="ltr" class="math inline">\(k&gt;0\)</span> הוא ב-s<span dir="ltr" class="math inline">\(^{-1}\)</span>; הפרשי טמפרטורה ניתנים למדידה בקלווין או במעלות צלזיוס. אנו מניחים שאיבוד החום פרופורציונלי בקירוב להפרש זה, במקום לטעון שהמודל מכסה אידוי או השפעות קרינה גדולות.</p>
<p dir="rtl" style="text-align:right">המחברת של התרגול מספקת את העדכון הפיסיקלי. אנו מזהים אתחול, תנאי, עדכון ומונה, ואז משתמשים בלולאת <code dir="ltr">while</code> כדי למצוא את הדגימה הראשונה על הסף או מתחתיו. שוויון נחשב הגעה לסף. אנו חוזים מה קורה אם הדגימה הראשונית כבר מתאימה, ומדוע סף מתחת לטמפרטורת החדר אינו ניתן להשגה על ידי קירור מונוטוני מלמעלה. אחרי דיון של 20 דקות אנו מתחילים את תרגיל הבית של פטיש, שהלולאה שלו עוצרת בדגימה הראשונה מתחת לקרקע.</p>
<h2 id="תרגיל-בית-ראשי-זריקת-פטיש-עם-התנגדות-אוויר">תרגיל בית ראשי: זריקת פטיש עם התנגדות אוויר</h2>
<p dir="rtl" style="text-align:right">אנו מחקים זריקת פטיש דו-ממדית מהראשית עם התנגדות אוויר ריבועית. המודל ליחידת מסה הוא <span dir="ltr" class="math display">\[\dot x=v_x,\quad \dot y=v_y,\quad \dot v_x=-b v v_x,\quad
\dot v_y=-g-b v v_y,\qquad v=\sqrt{v_x^2+v_y^2},\]</span> כאשר <span dir="ltr" class="math inline">\(g=9.81\)</span> m/s<span dir="ltr" class="math inline">\(^2\)</span> ומקדם גרר <span dir="ltr" class="math inline">\(b\approx0.001\)</span> m<span dir="ltr" class="math inline">\(^{-1}\)</span>. התוכנית קוראת את המהירות ההתחלתית, זווית הזריקה במעלות, הזמן הכולל <span dir="ltr" class="math inline">\(T\)</span>, צעד הזמן <span dir="ltr" class="math inline">\(\Delta t\)</span>, ואת <span dir="ltr" class="math inline">\(b\)</span>, מקדמת את המצב עם אוילר מפורש ומאחסנת היסטוריות מיושרות של <span dir="ltr" class="math inline">\(x,y,v_x,v_y\)</span>. היא עוצרת בדגימה המוצעת הראשונה עם <span dir="ltr" class="math inline">\(y&lt;0\)</span>, או אחרי <span dir="ltr" class="math inline">\(T\)</span>, ומדפיסה את המרחק האופקי שהושג.</p>
<p dir="rtl" style="text-align:right">תחילה נשרטט את המסלול הצפוי ונציין יחידות וסימנים. אחר כך נבנה את הווקטורים ואת הלולאה, תוך הקפדה על עדכון מקום מהמהירות הישנה. נבדוק את העדכון הראשון ביד ונאמת שכל וקטור גדל באיבר אחד לכל צעד, שהמהירות האופקית דועכת רק דרך גרר, ושהתאוצה האנכית כוללת גם כבידה וגם את איבר הגרר. נריץ את התוכנית עם הדוגמה הנתונה, נשווה את המרחק המודפס לאומדן שחושב ביד, ונעדן את <span dir="ltr" class="math inline">\(\Delta t\)</span> כדי לראות כיצד המרחק הנדגם זז. נשמור את גרף המסלול כמוצר. הדו"ח חייב לציין איזה כלל עצר את הלולאה: חציית הקרקע או סוף חלון הסימולציה. ההרחבה האופציונלית בגיליון, הבודקת האם <span dir="ltr" class="math inline">\(45^\circ\)</span> עדיין ממקסמת את הטווח, אינה נדרשת. הגדרת פונקציות נותרת חומר של שבוע 5.</p>
<h2 id="בונוס-אופציונלי-שלוש-סכמות-אינטגרציה">בונוס אופציונלי: שלוש סכמות אינטגרציה</h2>
<p dir="rtl" style="text-align:right">בונוס זה הוא אופציונלי, אינו תנאי מוקדם לעבודה מאוחרת יותר, ואינו נבחן. הוא שווה <strong>שליש נקודה בציון הסופי</strong>.</p>
<p dir="rtl" style="text-align:right">אנו משווים אוילר קדמי, אוילר-קרומר, וסכמת velocity-Verlet (נקודת אמצע) על נפילה חופשית באמצעות מעקב אחר האנרגיה המכנית ליחידת מסה. אנו קוראים את הזמן הסופי, צעד הזמן, הגובה ההתחלתי והמהירות ההתחלתית, מריצים את שלוש הסכמות על אותה רשת ומדפיסים את שלוש האנרגיות הסגוליות הסופיות. לאחר מכן נשרטט את האנרגיה הסגולית כפונקציה של הזמן עבור שלוש הסכמות יחד עם הערך השמור המדויק. שתי שאלות תיאורטיות קצרות מבקשות מאתנו לגזור את שינוי האנרגיה לכל צעד עבור כל סכמה ולומר איזו סכמה עדיפה ומדוע.</p>
<h2 id="המשך-חקר-מהליכה-אחת-לדיפוזיה">המשך חקר: מהליכה אחת לדיפוזיה</h2>
<p dir="rtl" style="text-align:right">מסלול אחד יכול לנדוד חזרה לכיוון המוצא. דיפוזיה היא הצהרת אנסמבל: עבור צעדים יחידתיים בלתי תלויים ונטולי הטיה, ההעתק הריבועי הממוצע מקיים <span dir="ltr" class="math inline">\(\langle r_n^2\rangle=n\)</span> ביחידות סריג, כי איברי ההצלבה מתאפסים בממוצע. אם צעד אחד נמשך <span dir="ltr" class="math inline">\(\tau\)</span> ואורכו <span dir="ltr" class="math inline">\(\ell\)</span>, מקדם הדיפוזיה הדו-ממדי המקביל הוא <span dir="ltr" class="math inline">\(D=\ell^2/(4\tau)\)</span>.</p>
<p dir="rtl" style="text-align:right">באמצעות מתכון אנסמבל מסופק, אנו משווים אוסף קטן וגדול יותר של הליכות, משרטטים העתק ריבועי ממוצע כנגד מספר הצעד, ובוחנים תנודות סביב הקו הישר הצפוי. אנו מסבירים מדוע הריבוע של ההעתק הממוצע הוא גודל שונה ויכול להישאר קרוב לאפס. אנו מתעדים את הגדרת הגנרטור ואת גודל האנסמבל עם הגרף. חקר זה הוא אופציונלי ואינו נבחן; מכונות האנסמבל המסופקת אינה תנאי מוקדם לעבודה ראשית.</p>
