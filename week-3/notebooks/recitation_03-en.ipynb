## Instructor-led problem (at most 20 minutes)

#### Sampled cooling

We model a body whose temperature excess above a fixed environment is multiplied by $0.9$ each minute. In general $T_{n+1}=T_{\rm room}+0.9(T_n-T_{\rm room})$. We choose $T_{\rm room}=0^\circ\mathrm C$ for this example, so $T_{n+1}=0.9T_n$. This corresponds to exact samples of Newton cooling with rate $-\log(0.9)$ per minute. The model assumes a uniform body temperature and an approximately linear heat-loss law; multiplying an arbitrary Celsius temperature by 0.9 without specifying the environment would be physically misleading. We seek the first sample at or below the threshold; equality counts.

**Before running:** predict the value, printed text, or graph described above. This is the corresponding example from the written chapter.

````julia
begin temperature = 100.0   # C
threshold = 81.0          # C
minute = 0
while temperature > threshold
    temperature = 0.9 * temperature
    minute = minute + 1
end
println((minute, temperature))
end
````

## Going further: from one walk to diffusion

One path can wander back toward its origin. Diffusion is an ensemble statement: for independent unbiased unit steps, the mean-square displacement satisfies $\langle r_n^2\rangle=n$ in lattice units, since cross terms average to zero. If one step lasts $\tau$ and has length $\ell$, the corresponding two-dimensional diffusion coefficient is $D=\ell^2/(4\tau)$.
