<h2 id="השאלה-הפיזיקלית">1.1 השאלה הפיזיקלית</h2>
<p>נדמיין מסה קטנה המחוברת לקפיצים בשני כיוונים ניצבים, ונצפה בה מלמעלה.
כל קואורדינטה יכולה להתנודד באופן סינוסואידלי, ובכל זאת המסלול המשולב
עשוי להיות אליפסה, קו ישר או תבנית מחזורית מורכבת. כיצד תמונה דו־ממדית
מלמדת אותנו על שני שעונים?</p>
<p>אותה מתמטיקה מתארת את ההסטה האופקית והאנכית של אוסצילוסקופ במצב XY.
שם הקואורדינטות מייצגות מתחים ולא אורכים. לגרף של אות אחד כנגד האחר יש
פרמטר זמן משותף, אף שהזמן אינו ציר בגרף. נבחין בין <strong>עקומה
פרמטרית</strong> זו לבין שני גרפים של מקום כתלות בזמן.</p>
<blockquote class="physics" style="background:#f4eff9;border:0;border-inline-start:5px solid #8c69ab;padding:.5em 1.2em;margin:1.5em 0">
<p><strong>פיזיקה — שני כוחות מחזירים בלתי תלויים.</strong> נבחר את
הכיוון ימינה כחיובי בציר <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(x\)</span> ואת
הכיוון למעלה כחיובי בציר <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(y\)</span>. למסה
<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(m\)</span> בק״ג פועלים <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(F_x=-k_xx\)</span> ו־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(F_y=-k_yy\)</span>, כאשר <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(k_x,k_y&gt;0\)</span> הם קבועי קפיץ ביחידות N/m.
בהנחת העתקות קטנות, קפיצים ליניאריים, חיכוך זניח וללא צימוד בין הכיוונים
מתקבלות המשוואות <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(m\ddot{x}=-k_xx\)</span>
ו־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(m\ddot{y}=-k_yy\)</span>. סימן המינוס הופך
את הכוח למחַזיר. הראשית היא שיווי המשקל; אין זה בדרך כלל מסלול כבידתי.
כשההעתק בכיוון מסוים אפס, הכוח באותו כיוון אפס. העתקות גדולות, חיכוך,
אילוץ חיצוני או צימוד מחייבים מודל אחר.</p>
</blockquote>
<p>הנגזרת השנייה של קוסינוס היא מינוס אותה פונקציה כפול ריבוע התדירות
הזוויתית. הצבה במשוואת התנועה נותנת <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\omega_x=\sqrt{k_x/m}\)</span>. נבחר את ראשית הזמן
כך שלתנודת <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(y\)</span> יהיה מופע סינוס
אפס:</p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math display">\[x(t)=A\cos(\omega_xt+\delta),\qquad
y(t)=B\sin(\omega_yt).\]</span></p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(A,B\)</span> הן משרעות חיוביות במטרים,
<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(t\)</span> בשניות, <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\omega_x,\omega_y\)</span> תדירויות זוויתיות
חיוביות ב־rad/s, ו־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\delta\)</span> ברדיאנים.
מחזור שלם הוא <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(2\pi\)</span> רדיאנים. התדירות
הרגילה היא <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(f=\omega/(2\pi)\)</span> בהרץ
וזמן המחזור <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(T=1/f\)</span>. נתחיל ב־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(A=0.02\)</span> מ׳ וב־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(B=0.01\)</span> מ׳. הגדלת אחת המשרעות אמורה למתוח
את התמונה בלי לשנות את התזמון.</p>
<blockquote class="mathematics" style="background:#eff6ee;border:0;border-inline-start:5px solid #5f9763;padding:.5em 1.2em;margin:1.5em 0">
<p><strong>מתמטיקה — מופע הוא מיקום בתוך מחזור.</strong> הוספת <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(2\pi\)</span> למופע אינה משנה את הערך הנצפה.
במוסכמת הקוסינוס והסינוס שלנו, <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\delta=0\)</span> <strong>אינו</strong> אומר ששתי
הקואורדינטות באותו מופע: הקוסינוס כבר מקדים את הסינוס ברבע מחזור. לפני
פירוש סימן המופע, נכתוב את ההגדרות המדויקות.</p>
</blockquote>
<h2 id="מרגע-יחיד-לתנועה-שלמה">1.2 מרגע יחיד לתנועה שלמה</h2>
<p>נטען את חבילת השרטוט שכבר הותקנה. <code dir="ltr" style="unicode-bidi:isolate">using</code> מאפשר שימוש
בכליה, <code dir="ltr" style="unicode-bidi:isolate">gr()</code> בוחר מנוע גרפי ו־<code dir="ltr" style="unicode-bidi:isolate">default</code> קובע
הגדרות תצוגה. <code dir="ltr" style="unicode-bidi:isolate">Printf</code> מאפשר פלט מעוצב: <code dir="ltr" style="unicode-bidi:isolate">%.2f</code>
מציג שתי ספרות אחרי הנקודה בלי לעגל את המשתנה בזיכרון. הגדרות אלו
משפיעות על התצוגה, לא על הפיזיקה.</p>

</div>

```julia
using Plots
using Printf
gr()
default(size = (760, 420), linewidth = 2, legend = :topright)
println("Plots ready; coordinates in metres, time in seconds.")
```

<div dir="rtl">
<p>לפני ההרצה נחשב בראש: אחרי רבע שנייה בתדירות של הרץ אחד, הקוסינוס אפס
והסינוס אחד. לכן נצפה ל־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(x=0\)</span>
ול־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(y=B\)</span>.</p>

</div>

```julia
A = 0.02
B = 0.01
omega_x = 2pi
omega_y = 2pi
delta = 0.0
t = 0.25
x = A * cos(omega_x * t + delta)
y = B * sin(omega_y * t)
@printf("t = %.2f s: x = %.5f m, y = %.5f m\n", t, x, y)
```

<div dir="rtl">
<p>האפס המודפס מעוגל. בחישוב נקודה צפה הקוסינוס עשוי להיות בערך <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(6\times10^{-17}\)</span> ולא אפס מדויק. זהו דיוק
חישובי סופי של זהות מתמטית, ולא העתק פיזיקלי מדיד.</p>
<blockquote class="julia" style="background:#f0f1f3;border:0;border-inline-start:5px solid #858a93;padding:.5em 1.2em;margin:1.5em 0">
<p><strong>Julia — מילון שרטוט קצר.</strong> הביטוי
<code dir="ltr" style="unicode-bidi:isolate">range(0.0, 1.0; length = 501)</code> נותן 501 זמנים במרווחים
שווים, כולל הקצוות. <code dir="ltr" style="unicode-bidi:isolate">length</code> סופר אותם ו־<code dir="ltr" style="unicode-bidi:isolate">step</code>
מחזיר את המרווח. וקטור מכיל ערך לכל זמן. הנקודות
ב־<code dir="ltr" style="unicode-bidi:isolate">cos.(...)</code>, ב־<code dir="ltr" style="unicode-bidi:isolate">.*</code> וב־<code dir="ltr" style="unicode-bidi:isolate">.+</code> אומרות
להפעיל את הפעולה על כל איבר בנפרד. למשל ריבוע איברי הווקטור 2, 3 נותן 4,
9, ולא מכפלת מטריצות. <code dir="ltr" style="unicode-bidi:isolate">plot(times, xs)</code> מצמיד זמן ומיקום בעלי
אותו אינדקס. <code dir="ltr" style="unicode-bidi:isolate">plot!</code> מוסיף סדרה לאיור קיים.
<code dir="ltr" style="unicode-bidi:isolate">display</code> מציג איור במפורש, גם בהרצה כקובץ תוכנית.</p>
</blockquote>

</div>

```julia
times = range(0.0, 1.0; length = 501)
xs = A .* cos.(omega_x .* times .+ delta)
ys = B .* sin.(omega_y .* times)
@printf("%d samples; dt = %.4f s\n", length(times), step(times))
p_time = plot(times, xs; label = "x(t)", xlabel = "t (s)", ylabel = "position (m)")
plot!(p_time, times, ys; label = "y(t)")
display(p_time)
p_orbit = plot(xs, ys; label = "equal frequencies", xlabel = "x (m)",
    ylabel = "y (m)", aspect_ratio = :equal)
display(p_orbit)
```

<div dir="rtl">
<p>באיור הראשון רואים את שני השעונים. בשני הזמן הוסר מן הצירים, ואנו
עוקבים אחרי זוג הקואורדינטות. יחס צירים שווה חשוב: למטר אופקי ולמטר אנכי
אותו אורך חזותי. כאשר <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(A\ne B\)</span> נקבל
אליפסה ולא מעגל. הקווים מחברים דגימות של מודל אנליטי; אלו אינן מדידות
חדשות או תוצאות של פותר משוואות דיפרנציאליות.</p>
<h2 id="גזירת-האליפסה-בתדירויות-שוות">1.3 גזירת האליפסה בתדירויות
שוות</h2>
<p>נסמן <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(u=\omega t\)</span>, <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(X=x/A\)</span> ו־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(Y=y/B\)</span>. נוסחת חיבור הזוויות נותנת</p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math display">\[X=\cos u\cos\delta-\sin u\sin\delta,
\qquad X+Y\sin\delta=\cos u\cos\delta.\]</span></p>
<p>נעלה בריבוע ונשתמש ב־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\cos^2u=1-Y^2\)</span>:</p>
<p><span
style="direction:ltr;unicode-bidi:isolate" class="math display">\[\boxed{X^2+Y^2+2XY\sin\delta=\cos^2\delta.}\]</span></p>
<p>עבור <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\delta=0\)</span> נקבל <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(X^2+Y^2=1\)</span>. עבור <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\delta=\pi/2\)</span> נקבל <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\((X+Y)^2=0\)</span>, כלומר את הישר <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(X=-Y\)</span>. ב־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\delta=-\pi/2\)</span> השיפוע חיובי. במופעי ביניים
נקבל אליפסה נטויה. אלה תוצאות גאומטריות של המופע, ולא עדות לאובדן
אנרגיה.</p>
<p>ננבא את שתי הצורות באיור הבא. השארית היא צד שמאל פחות צד ימין של
המשוואה במסגרת, בכל נקודת דגימה. <code dir="ltr" style="unicode-bidi:isolate">abs.</code> לוקח ערכים מוחלטים
ו־<code dir="ltr" style="unicode-bidi:isolate">maximum</code> מוצא את הגדול שבהם; גם העלאה בחזקה עם נקודה פועלת
איבר־איבר. שארית זעירה בודקת יחד את הגזירה ואת הקוד.</p>

</div>

```julia
delta = pi / 4
xs = A .* cos.(omega_x .* times .+ delta)
ys = B .* sin.(omega_y .* times)
X = xs ./ A
Y = ys ./ B
ellipse_residual = X .^ 2 .+ Y .^ 2 .+ 2 .* X .* Y .* sin(delta) .- cos(delta)^2
@printf("Largest ellipse-equation residual: %.2e\n", maximum(abs.(ellipse_residual)))
p_phase = plot(xs, ys; label = "delta = pi/4", xlabel = "x (m)",
    ylabel = "y (m)", aspect_ratio = :equal)
plot!(p_phase, A .* cos.(omega_x .* times .+ pi / 2), ys; label = "delta = pi/2")
display(p_phase)
```

<div dir="rtl">
<p>הישר והאליפסה הנטויה תואמים את הגזירה. הספרות האחרונות של השארית
עשויות להשתנות בין ספריות חישוב, אך סדר הגודל צריך להיות קרוב לדיוק
המכונה. את הישר עוברים הלוך ושוב: צורת המסלול לבדה אינה מציגה את כיוון
התנועה או את המהירות.</p>
<h2 id="מדוע-תבניות-מסוימות-נסגרות">1.4 מדוע תבניות מסוימות נסגרות?</h2>
<p>כדי ש<strong>המצב המלא</strong> יחזור, גם המקומות וגם המהירויות
צריכים לחזור. עבור משרעות שאינן אפס ותדירויות חיוביות, התנאי הוא התקדמות
של שני המופעים במספר שלם של מחזורים:</p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math display">\[\omega_xT=2\pi p,\qquad \omega_yT=2\pi
q,\]</span></p>
<p>כאשר <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(p,q\)</span> שלמים חיוביים. מחלוקה
נובע שהיחס <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\omega_x/\omega_y=p/q\)</span>
רציונלי. אם אין ל־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(p,q\)</span> גורם משותף
ו־<span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\omega_x=p\Omega\)</span>, <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\omega_y=q\Omega\)</span>, זמן המחזור הבסיסי של
המצב הוא <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(2\pi/\Omega\)</span>.</p>
<p>למשל, <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(f_x=3\)</span> הרץ ו־<span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(f_y=2\)</span> הרץ נותנים חזרה לאחר שנייה אחת.
מגזירה נקבל</p>
<p><span
style="direction:ltr;unicode-bidi:isolate" class="math display">\[v_x=-A\omega_x\sin(\omega_xt+\delta),\qquad
v_y=B\omega_y\cos(\omega_yt).\]</span></p>
<p>נשווה את קצות חלון הזמן. <code dir="ltr" style="unicode-bidi:isolate">xs[1]</code> הוא האיבר הראשון
ו־<code dir="ltr" style="unicode-bidi:isolate">xs[end]</code> האחרון. <code dir="ltr" style="unicode-bidi:isolate">hypot(dx,dy)</code> מחשב <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\sqrt{dx^2+dy^2}\)</span>. הסכמה במיקום בלבד יכולה
לציין גם חצייה מקרית של המסלול.</p>

</div>

```julia
omega_x = 3 * 2pi
omega_y = 2 * 2pi
delta = 0.3
times = range(0.0, 1.0; length = 1501)
xs = A .* cos.(omega_x .* times .+ delta)
ys = B .* sin.(omega_y .* times)
vx = -A .* omega_x .* sin.(omega_x .* times .+ delta)
vy = B .* omega_y .* cos.(omega_y .* times)
@printf("Position closure error: %.2e m\n", hypot(xs[end] - xs[1], ys[end] - ys[1]))
@printf("Velocity closure error: %.2e m/s\n", hypot(vx[end] - vx[1], vy[end] - vy[1]))
display(plot(xs, ys; label = "3:2, T = 1 s", xlabel = "x (m)",
    ylabel = "y (m)", aspect_ratio = :equal))
```

<div dir="rtl">
<p>ההפרשים הקטנים הן במקום והן במהירות תומכים בזמן המחזור שחזינו. בחצייה
בתוך העקומה המהירויות בשני המעברים בדרך כלל שונות; זו אינה חזרה של כל
המצב הפיזיקלי.</p>
<blockquote class="check" style="background:#fff4e7;border:0;border-inline-start:5px solid #d98538;padding:.5em 1.2em;margin:1.5em 0">
<p><strong>בדיקה — סגירה אינה תהודה.</strong> תיארנו תנודות חופשיות ללא
כוח מאלץ מחזורי. יחס תדירויות רציונלי גורם לתמונה לחזור, אך אינו כשלעצמו
מעיד על קליטת אנרגיה בתהודה. לתיאור תנודה מאולצת צריך להוסיף כוח חיצוני
ובדרך כלל גם ריסון.</p>
</blockquote>
<h2 id="שעונים-כמעט-שווים-וסגירה-מדומה">1.5 שעונים כמעט שווים וסגירה
מדומה</h2>
<p>נבחר תדירויות 1.01 הרץ ו־1 הרץ. המופע היחסי משתנה בקצב <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\Delta\omega=0.01(2\pi)\)</span> rad/s, ולכן משלים
מחזור בזמן</p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math display">\[T_{\rm
relative}=\frac{2\pi}{|\Delta\omega|}=100\ \mathrm{s}.\]</span></p>
<p>בשנייה אחת נראה כמעט אליפסה. במשך מאה שניות מצטיירים גדילים רבים.
ובכל זאת 1.01 הוא המספר הרציונלי 101/100 במודל המתמטי, ולכן הדוגמה נסגרת
לבסוף. <code dir="ltr" style="unicode-bidi:isolate">layout = (1, 2)</code> מציג שורה אחת ושתי עמודות של
איורים.</p>

</div>

```julia
omega_x = 1.01 * 2pi
omega_y = 2pi
delta = 0.0
short_t = range(0.0, 1.0; length = 1001)
long_t = range(0.0, 100.0; length = 100001)
@printf("Relative-phase cycle: %.1f s\n", 2pi / abs(omega_x - omega_y))
p_short = plot(A .* cos.(omega_x .* short_t), B .* sin.(omega_y .* short_t);
    label = "first second", xlabel = "x (m)", ylabel = "y (m)", aspect_ratio = :equal)
p_long = plot(A .* cos.(omega_x .* long_t), B .* sin.(omega_y .* long_t);
    label = "100 seconds", xlabel = "x (m)", ylabel = "y (m)", aspect_ratio = :equal)
display(plot(p_short, p_long; layout = (1, 2), size = (960, 400)))
```

<div dir="rtl">
<p>ליחס תדירויות אי־רציונלי מדויק אין זמן חזרה סופי של המצב. בתנועה
אידיאלית כזאת הנקודות צפופות במלבן המותר לאורך זמן ארוך כרצוננו, אף שכל
קטע סופי הוא רק עקומה. ציור צפוף ותדירויות עשרוניות בדיוק סופי אינם
הוכחה לאי־רציונליות. גם תדירויות מדודות אינן ידועות בדיוק מוחלט. אפשר
לדווח על חזרה בטווח סבילות ובחלון זמן מוגדרים; ניסוי סופי אינו מוכיח
טענה על זמן אינסופי.</p>
<h2 id="האם-השרטוט-יכול-להמציא-צורה">1.6 האם השרטוט יכול להמציא
צורה?</h2>
<p>המחשב מחבר נקודות בקווים ישרים. ארבעה מרווחים סביב אליפסה יוצרים
מצולע דמוי מעוין. משוואת התנועה לא השתנתה; רק הייצוג השתנה.</p>

</div>

```julia
coarse_t = range(0.0, 1.0; length = 5)
fine_t = range(0.0, 1.0; length = 1001)
p_alias = plot(A .* cos.(2pi .* fine_t), B .* sin.(2pi .* fine_t);
    label = "dense model", xlabel = "x (m)", ylabel = "y (m)", aspect_ratio = :equal)
plot!(p_alias, A .* cos.(2pi .* coarse_t), B .* sin.(2pi .* coarse_t);
    marker = :circle, label = "4 intervals")
display(p_alias)
```

<div dir="rtl">
<blockquote class="check" style="background:#fff4e7;border:0;border-inline-start:5px solid #d98538;padding:.5em 1.2em;margin:1.5em 0">
<p><strong>בדיקה — דגימה וקיפול תדרים.</strong> כדי למנוע את עמימות
הדגימה הפשוטה של סינוס, תדירות הדגימה צריכה להיות גדולה מפעמיים תדירותו.
תנאי זה לבדו אינו מבטיח איור חלק ומועיל. נתחיל בלפחות 50–100 דגימות
למחזור המהיר ביותר ואז נחצה את מרווח הזמן. שינוי גדול בגאומטריה מעיד
שהדגימה הקודמת לא הספיקה. דגימה פעם אחת בכל מחזור יכולה אף להציג תנועה
כאילו היא קפואה.</p>
</blockquote>
<h2 id="אנרגיה-כבדיקה-פיזיקלית-נוספת">1.7 אנרגיה כבדיקה פיזיקלית
נוספת</h2>
<p>האנרגיה הפוטנציאלית היא <span style="direction:ltr;unicode-bidi:isolate" class="math inline">\(U=\tfrac12
k_xx^2+\tfrac12 k_yy^2\)</span>. יחד עם האנרגיה הקינטית נקבל</p>
<p><span style="direction:ltr;unicode-bidi:isolate" class="math display">\[E=\frac m2(v_x^2+v_y^2)+\frac
m2(\omega_x^2x^2+\omega_y^2y^2)
=\frac m2(\omega_x^2A^2+\omega_y^2B^2).\]</span></p>
<p>השוויון נובע בנפרד בכל כיוון מן הזהות <span
style="direction:ltr;unicode-bidi:isolate" class="math inline">\(\sin^2+\cos^2=1\)</span>. האנרגיה קבועה גם במסלול
מורכב או שאינו נסגר. בבדיקות הנתונות <code dir="ltr" style="unicode-bidi:isolate">@test</code> מציין תנאי שאמור
להתקיים, <code dir="ltr" style="unicode-bidi:isolate">isapprox</code> בודק הסכמה עד סבילות מוחלטת
ו־<code dir="ltr" style="unicode-bidi:isolate">@testset</code> מקבץ את התוצאות. הבדיקות משתמשות בווקטורי דוגמת
3:2 שנשמרו קודם; שינוי התדירויות המאוחר לא שינה אותם.</p>

</div>

```julia
using Test
# The final semicolon hides the returned test object; its summary still prints.
@testset "Oscillator geometry" begin
    @test isapprox(x, 0.0; atol = 1e-16)
    @test isapprox(y, B; atol = 1e-16)
    @test maximum(abs.(ellipse_residual)) < 1e-12
    @test hypot(xs[end] - xs[1], ys[end] - ys[1]) < 1e-12
    @test hypot(vx[end] - vx[1], vy[end] - vy[1]) < 1e-12
    m = 0.1
    energy = 0.5 .* m .* (vx .^ 2 .+ vy .^ 2) .+
        0.5 .* m .* ((3 * 2pi)^2 .* xs .^ 2 .+ (2 * 2pi)^2 .* ys .^ 2)
    expected_energy = 0.5 * m * ((3 * 2pi)^2 * A^2 + (2 * 2pi)^2 * B^2)
    @test maximum(abs.(energy .- expected_energy)) < 1e-12
end;
```

<div dir="rtl">
<p>הצלחה בודקת התאמה למודל האידיאלי, לא את נכונותו עבור קפיץ ממשי. דעיכת
משרעת מדודה תציע להוסיף ריסון; תלות של התדירות במשרעת תציע כוח מחזיר לא
ליניארי.</p>
