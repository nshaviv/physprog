<h4 id="يجب-أن-تنقل-الأسماء-المعنى">يجب أن تنقل الأسماء المعنى</h4>
<p dir="rtl" style="text-align:right"><strong>الأسماء</strong>: أسماء المتغيرات حسّاسة لحالة الأحرف، ولا يمكن أن تحتوي فراغات، ويجب استخدامها باتساق. ويمكن استخدام <code dir="ltr">unicode</code>، ولا سيّما الرموز اليونانية (مثل كتابة <code dir="ltr">\phi&lt;Tab&gt;</code> للحصول على <span dir="ltr" class="math inline">\(\phi\)</span>). نفضّل الأسماء ذات المعنى على أسماء غامضة مثل <code dir="ltr">a</code> و<code dir="ltr">b</code> و<code dir="ltr">c</code>. <strong>الوحدات</strong>: تخزّن <code dir="ltr">Julia</code> عادة أعدادًا عادية دون وحداتها. لكن في الفيزياء يُستحسن تتبّعها. ويمكن إضافة لواحق وحدية إلى الأسماء، مثل <code dir="ltr">m_kg</code>، أو بدلًا من ذلك، كما نفعل هنا، تسجيل الوحدة في تعليق قصير في نهاية السطر حيث تُعرَّف القيمة، مثل <code dir="ltr">time = 0.35 # s</code> و<code dir="ltr">height = 1.20 # m</code> و<code dir="ltr">g = 9.81 # m/s^2</code>؛ وهذا ليس إلزاميًا لكنه مفيد جدًا لتجنّب أخطاء الوحدات. وفي واجب المكافأة 2 (أي أنه ليس جزءًا من مادة المقرر الرسمية) سنرى أننا نستطيع استخدام حزمة <code dir="ltr">Unitful</code> التي تضيف وحدات إلى المتغيرات وتمكّننا من تتبّعها. ويجب أيضًا ذكر الوحدات في توثيق الدوال أو توثيق الواجهة، وتوضيحها عند مواقع الاستدعاء عندما تكون ملتبسة.</p>

<h2 id="الحركية-متعددة-القطع-كقرار">الحركية متعددة القطع كقرار</h2>
<h4 id="التسارع-الموجب-مقابل-مقدار-الجاذبية-الموجب">التسارع الموجَّب مقابل مقدار الجاذبية الموجب</h4>
<p dir="rtl" style="text-align:right">الارتفاع موجب نحو الأعلى، لذا يكون تسارع الجاذبية <span dir="ltr" class="math display">\[g=-9.81\ \mathrm{m/s^2}.\]</span> ولارتفاع ابتدائي <span dir="ltr" class="math inline">\(z_0\)</span> وسرعة <span dir="ltr" class="math inline">\(v_0\)</span>، <span dir="ltr" class="math display">\[z_{\mathrm{free}}(t)=z_0+v_0t+\frac{1}{2}gt^2.\]</span> وللحدود الثلاثة وحدات المتر. وهذا الفحص البُعدي مستقل عن كون قيمها في الكود من النوع <code dir="ltr">Float64</code>.</p>
<p dir="rtl" style="text-align:right">نموذجنا دون ارتداد لا يسمح للكرة بمواصلة الحركة تحت الأرض: <span dir="ltr" class="math display">\[z(t)=\begin{cases}
z_{\mathrm{free}}(t),&amp;z_{\mathrm{free}}(t)&gt;0,\\
0,&amp;z_{\mathrm{free}}(t)\leq0.
\end{cases}\]</span></p>
<p dir="rtl" style="text-align:right">ويتحوّل التعريف متعدد القطع مباشرةً إلى شرطية:</p>
</div>

<div dir="rtl" style="text-align:right">

</div>

```julia
t = 0.35   # s
z0 = 1.20   # m
v0 = 0.0    # m/s
g = -9.81   # m/s^2

if t < 0 || z0 < 0
    throw(ArgumentError("time and initial height must be non-negative"))
end

z_free = z0 + v0 * t + 0.5 * g * t^2   # m

if z_free > 0
    z = z_free
    state = "in air"
else
    z = 0.0
    state = "on the ground"
end

println("at t = $(t) s: z = $(round(z, digits=3)) m ($(state))")
```

<h2 id="نموذج-احتكاك-متعدد-القطع-على-منحدر">نموذج احتكاك متعدد القطع على منحدر</h2>
<h4 id="القوى-والإشارات-والافتراضات">القوى والإشارات والافتراضات</h4>
<p dir="rtl" style="text-align:right">نأخذ <span dir="ltr" class="math inline">\(+x\)</span> نحو الأعلى بمحاذاة منحدر ثابت و<span dir="ltr" class="math inline">\(+y\)</span> عموديًا عليه. ولجسم كتلته <span dir="ltr" class="math inline">\(m\)</span> (كغ) وزاوية <span dir="ltr" class="math inline">\(\theta\)</span> (بالراديان في <code dir="ltr">sin</code>/<code dir="ltr">cos</code>) ومقدار جاذبية موجب <span dir="ltr" class="math inline">\(g\)</span> (m/s<span dir="ltr" class="math inline">\(^2\)</span>)، للجاذبية مركّبة نزولية <span dir="ltr" class="math inline">\(mg\sin\theta\)</span> وقوة عمودية <span dir="ltr" class="math inline">\(N=mg\cos\theta\)</span>. ويمكن للاحتكاك السكوني أن يعارض الحركة الوشيكة نحو الأسفل حتى <span dir="ltr" class="math inline">\(f_{s,\max}=\mu_sN\)</span>. وإذا لم يستطع التثبيت، يعمل الاحتكاك الحركي <span dir="ltr" class="math inline">\(f_k=\mu_kN\)</span> نحو الأعلى ويكون <span dir="ltr" class="math inline">\(a=(-mg\sin\theta+\mu_kN)/m\)</span>. نفترض احتكاك كولوم وتلامسًا صلبًا ولا مقاومة هواء و<span dir="ltr" class="math inline">\(0\le\theta&lt;90^\circ\)</span>. ولا يصف النموذج التدحرج أو تغيّر التلامس أو احتكاكًا يعتمد على السرعة. وعند <span dir="ltr" class="math inline">\(\theta=0\)</span> يكون الاحتكاك السكوني المطلوب صفرًا؛ وإذا كان <span dir="ltr" class="math inline">\(\mu_k=0\)</span>، يعطي الانزلاق <span dir="ltr" class="math inline">\(a=-g\sin\theta\)</span>.</p>
<p dir="rtl" style="text-align:right">نتحقق قبل استخدام أي مدخل. يوقف <code dir="ltr">throw(ArgumentError(...))</code> الحساب فورًا ويخبر مستدعيه أن شرطًا موثّقًا في المدخل قد فشل. وهو يختلف عن <code dir="ltr">println</code> الذي يكتفي بعرض نص ثم يكمل. ويسأل اختبار العضوية <code dir="ltr">value isa T</code> هل تنتمي القيمة إلى النوع <code dir="ltr">T</code>؛ ويسمّي <code dir="ltr">Real</code> عائلة <code dir="ltr">Julia</code> للأنواع الحقيقية، التي تضم الأعداد الصحيحة والعائمة العادية. ويحتاج <code dir="ltr">Bool</code> إلى عناية خاصة هنا: فبما أن <code dir="ltr">true isa Real</code> صحيحة أيضًا، فإن فحص <code dir="ltr">Real</code> وحده سيقبل <code dir="ltr">true</code> ككتلة، لذا نرفضه صراحةً.</p>
</div>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>قبل التنفيذ:</strong> نتوقّع القيمة أو النص أو الرسم الموصوف أعلاه. هذا هو المثال المقابل من الفصل المكتوب.</p>
</div>

```julia
mass = 2.0      # kg
angle = 30.0    # deg
 μ_s = 0.60
 μ_k = 0.40
g = 9.81        # m/s^2
if !(mass isa Real) || mass isa Bool || !isfinite(mass) || mass <= 0
    throw(ArgumentError("mass must be finite, positive, and non-Boolean"))
elseif !isfinite(angle) || !(0 <= angle < 90) || μ_k < 0 || μ_k > μ_s
    throw(ArgumentError("use 0 <= angle < 90 and 0 <= μ_k <= μ_s"))
end
 θ = deg2rad(angle)   # rad
normal = mass * g * cos(θ)   # N
needed_static = mass * g * sin(θ)   # N
if needed_static <= μ_s * normal
    state = "stuck"
    acceleration = 0.0   # m/s^2
else
    state = "sliding"
    acceleration = (-needed_static + μ_k * normal) / mass   # m/s^2
end
println(state, ", a = ", acceleration, " m/s^2")
```

<h2 id="التمرين-قوة-سحب-قد-تعكس-حركة-وشيكة">التمرين: قوة سحب قد تعكس حركة وشيكة</h2>
<p dir="rtl" style="text-align:right">نضيف إلى نموذج المنحدر قوة سحب معروفة نحو الأعلى <span dir="ltr" class="math inline">\(P\)</span>. وقبل اختيار إشارة الاحتكاك، نحسب النزعة بلا احتكاك <span dir="ltr" class="math inline">\(T=P-mg\sin\theta\)</span>. فالنزعة الموجبة <span dir="ltr" class="math inline">\(T\)</span> تعني حركة وشيكة نحو الأعلى، والسلبية تعني حركة وشيكة نحو الأسفل. ويعارض الاحتكاك تلك النزعة، ويمكن للاحتكاك السكوني أن يثبّت متى كان <span dir="ltr" class="math inline">\(|T|\le\mu_sN\)</span>.</p>
<p dir="rtl" style="text-align:right">نصل قادرين على قراءة سلسلة <code dir="ltr">if</code>/<code dir="ltr">elseif</code>/<code dir="ltr">else</code> وتحويل الدرجات إلى راديان. وفي الجزء الموجَّه القصير نتوقّع الاتجاه، ونتتبّع فرعًا واحدًا، ونختبر <span dir="ltr" class="math inline">\(T=0\)</span>. ثم نبدأ واجب خلوص الجدار المستقل. والفكرة الجوهرية طريقة قابلة للتكرار: نتحقق من المدخلات، ونصوغ شرطًا فيزيائيًا، ونحسب الحالة المنطبقة، ونفحص حدّها.</p>
<h2 id="الواجب-الأساسي-هل-يعبر-المقذوف-جدارا">الواجب الأساسي: هل يعبر المقذوف جدارًا؟</h2>
<p dir="rtl" style="text-align:right">نطلق من مستوى الأرض بسرعة <span dir="ltr" class="math inline">\(v_0\)</span> وزاوية <span dir="ltr" class="math inline">\(\theta\)</span> ومقدار جاذبية <span dir="ltr" class="math inline">\(g&gt;0\)</span>، مع <span dir="ltr" class="math inline">\(+x\)</span> نحو الجدار و<span dir="ltr" class="math inline">\(+y\)</span> نحو الأعلى. وفي نموذج الجاذبية المنتظمة دون مقاومة هواء، <span dir="ltr" class="math display">\[v_x=v_0\cos\theta,\quad t_w=x_w/v_x,\quad y_w=v_0\sin\theta\,t_w-\tfrac12gt_w^2,\quad v_{y,w}=v_0\sin\theta-gt_w.\]</span> ونشترط <span dir="ltr" class="math inline">\(v_x&gt;0\)</span> قبل القسمة. نقارن <span dir="ltr" class="math inline">\(y_w\)</span> بارتفاع الجدار، ونستخدم <span dir="ltr" class="math inline">\(v_{y,w}\)</span> لتصنيف الصعود أو الهبوط. والمساواة في الارتفاع تُحتسب عبورًا؛ والسرعة الرأسية الصفرية تنتمي إلى فئة الهبوط المنشورة. وقد يكون الارتفاع المحسوب في الطيران الحر سالبًا: وبالنسبة لارتفاع جدار غير سالب يعني ذلك فشل العبور، لا استمرار الحركة فيزيائيًا عبر الأرض.</p>
<p dir="rtl" style="text-align:right">نكتب أولًا تنبؤًا لحالة عادية وأخرى حدّية. ثم نطبّق الشيفرة العددية، ونتحقق من مجالها الموثّق، ونقارن التسميات والقيمة المنطقية بدقة. ونتحقق من الارتفاعات تقريبيًا بالتحمّل المنشور. وأخيرًا نشرح كيف يختلف قرار العبور عن قرار الحالة الرأسية: فقد يعبر المقذوف وهو هابط. وورقة الواجب هي المكان الذي تُذكر فيه أسماء المدخلات الدقيقة والاختبارات واصطلاح غلاف الإرجاع وتعليمات التسليم؛ ولا حاجة إلى دوال أو حلقات يعرّفها الطالب.</p>
<h2 id="مكافأة-اختيارية-الوحدات-وأطوار-التيار-المتردد">مكافأة اختيارية: الوحدات وأطوار التيار المتردد</h2>
<p dir="rtl" style="text-align:right">هذه المكافأة اختيارية، وليست شرطًا مسبقًا لعمل لاحق، وغير ممتحنة. وتساوي <strong>ثلث نقطة من العلامة النهائية</strong>.</p>
<p dir="rtl" style="text-align:right">تطوّر القراءة المستقلة الأعداد المركبة والمقدار والطور والمرافق وعلاقة أويلر قبل استخدام <code dir="ltr">cis</code>. نختار سعات الذروة واصطلاح الزمن <span dir="ltr" class="math inline">\(\exp(+i\omega t)\)</span>، ونضع طور الجهد صفرًا، ونبني <span dir="ltr" class="math inline">\(Z=|Z|\exp(i\phi_Z)\)</span>، ونحسب <span dir="ltr" class="math inline">\(I=V/Z\)</span>. والتيار القابل للقياس هو الجزء الحقيقي من <span dir="ltr" class="math inline">\(I\exp(i\omega t)\)</span>؛ فالعدد المركب تمثيل، لا تيار مقيس إضافي.</p>
<p dir="rtl" style="text-align:right">نتبع أمثلة <code dir="ltr">Unitful</code> المُعطاة، ونتحقق من أن قسمة الجهد على المقاومة لها وحدة التيار، ونفحص مقاومًا بطور صفري قبل تجربة طور ربع دورة. ثم نقدّم الشيفرة العددية التي تسند التيار المركب بالأمبير ونشغّل فحوص الوحدات المكافئة والمدخلات غير الصالحة المنشورة. ونشرح لماذا يعطي طور معاوقة موجب طور تيار سالبًا. هذا نموذج أحادي التردد خطي في الحالة المستقرة؛ والانتقالات العابرة والمكوّنات غير الخطية خارج نطاقه.</p>
<h2 id="استكشاف-إضافي-الانعكاس-الداخلي-الكلي">استكشاف إضافي: الانعكاس الداخلي الكلي</h2>
<p dir="rtl" style="text-align:right">نستخدم وصفة قانون سنيل المُعطاة <span dir="ltr" class="math inline">\(n_1\sin\theta_1=n_2\sin\theta_2\)</span>، مع قرائن انكسار موجبة وزوايا مقيسة من العمودي على السطح الفاصل. وعندما يكون <span dir="ltr" class="math inline">\(n_1&gt;n_2\)</span>، يعطي السقوط بعد <span dir="ltr" class="math inline">\(\theta_c=\arcsin(n_2/n_1)\)</span> انعكاسًا داخليًا كليًا. ونصنّف الأمثلة دون الحرجة والحرجة وفوق الحرجة قبل حساب جيب معكوس؛ فحساب <code dir="ltr">asin</code> لعدد أكبر من واحد دون تفكير ليس حسابًا فيزيائيًا.</p>
<p dir="rtl" style="text-align:right">نرسم انعكاسات مثالية متكررة في ليف مستقيم ونغيّر زاوية الإطلاق. ونسجّل الخيارات التي تبقى موجَّهة ونشرح العمودي المستخدم عند كل سطح فاصل. وأي حلقة تتبع مُعطاة هي معاينة اختيارية، لا عمل أساسي في الأسبوع الثاني. وتتطلب أنماط الدليل الموجي والتشتت وفقد الانحناء والامتصاص فيزياءً أكثر من نموذج الأشعة هذا. وهذا الاستقصاء منفصل عن مكافأة الطور، وهو غير ممتحن وغير شرط مسبق.</p>
