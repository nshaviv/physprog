## Physics reference

#### Pulled block

Take uphill as positive. A known pull $P$ competes with $mg\sin\theta$. The tendency without friction is $T=P-mg\sin\theta$. Static friction opposes that tendency: if $T>0$, it points downhill; if $T<0$, it points uphill. It may hold when $|T|\le\mu_smg\cos\theta$. If it cannot, kinetic friction still opposes the impending motion and $a=(T-\operatorname{sign}(T)\mu_kmg\cos\theta)/m$. We assume $0\le\mu_k\le\mu_s$ and $0\le\theta<90^\circ$, together with Coulomb friction, rigid contact, and a fixed incline. At $T=0$, no friction is required; a horizontal plane and zero friction provide limiting checks.

## Core homework: does a projectile clear a wall?

We launch from ground level with speed $v_0$, angle $\theta$, and gravity magnitude $g>0$, with $+x$ toward the wall and $+y$ upward. In the uniform-gravity, no-drag model, $$v_x=v_0\cos\theta,\quad t_w=x_w/v_x,\quad y_w=v_0\sin\theta\,t_w-\tfrac12gt_w^2,\quad v_{y,w}=v_0\sin\theta-gt_w.$$ We require $v_x>0$ before dividing. We compare $y_w$ with the wall height, and use $v_{y,w}$ to classify rising or falling. Equality in height counts as clearing; zero vertical velocity belongs to the published falling category. The computed free-flight height may be negative: for a nonnegative wall height this means failure to clear, not motion physically continuing through the ground.
