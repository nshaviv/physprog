<p dir="rtl" style="text-align:right">אפשר לבדוק משתנה על ידי כתיבת שמו בתא של מחברת, או להדפיס אותו כחלק מהודעה:</p>

<h4 id="שמות-צריכים-לבטא-משמעות">שמות צריכים לבטא משמעות</h4>
<p dir="rtl" style="text-align:right"><strong>שמות</strong>: שמות משתנים מבחינים בין אותיות גדולות לקטנות, אינם יכולים להכיל רווחים, ויש להשתמש בהם בעקביות. אפשר להשתמש ב-<code dir="ltr">unicode</code>, ובמיוחד באותיות יווניות (למשל להקליד <code dir="ltr">\phi&lt;Tab&gt;</code> כדי לקבל <span dir="ltr" class="math inline">\(\phi\)</span>). נעדיף שמות בעלי משמעות על פני שמות לא מוסברים כגון <code dir="ltr">a</code>, <code dir="ltr">b</code> ו-<code dir="ltr">c</code>. <strong>יחידות</strong>: <code dir="ltr">Julia</code> שומרת בדרך כלל מספרים רגילים בלי היחידות שלהם. בפיסיקה, לעומת זאת, כדאי לעקוב אחריהן. אפשר להוסיף סיומות יחידה לשמות, למשל <code dir="ltr">m_kg</code>, או לחלופין, כפי שאנו עושים כאן, לתעד את היחידה בהערה קצרה בסוף השורה במקום שבו הערך מוגדר, למשל <code dir="ltr">time = 0.35 # s</code>, <code dir="ltr">height = 1.20 # m</code> ו-<code dir="ltr">g = 9.81 # m/s^2</code>; זה אינו חובה אך שימושי מאוד כדי להימנע משגיאות יחידות. במטלת הבונוס 2 (כלומר, שאינה חלק מחומר הקורס הרשמי) נראה שאפשר להשתמש בחבילה <code dir="ltr">Unitful</code> שמוסיפה יחידות למשתנים ומאפשרת לעקוב אחריהן. יש לציין יחידות גם במחרוזות התיעוד או בתיעוד הממשק של פונקציות, ולהבהיר אותן באתרי קריאה שבהם הן עלולות להיות עמומות.</p>

<h2 id="קינמטיקה-בחלקים-כהחלטה">קינמטיקה בחלקים כהחלטה</h2>
<h4 id="תאוצה-מסומנת-לעומת-גודל-הכבידה-החיובי">תאוצה מסומנת לעומת גודל הכבידה החיובי</h4>
<p dir="rtl" style="text-align:right">גובה חיובי כלפי מעלה, ולכן תאוצת הכובד היא <span dir="ltr" class="math display">\[g=-9.81\ \mathrm{m/s^2}.\]</span> עבור גובה התחלתי <span dir="ltr" class="math inline">\(z_0\)</span> ומהירות התחלתית <span dir="ltr" class="math inline">\(v_0\)</span>, <span dir="ltr" class="math display">\[z_{\mathrm{free}}(t)=z_0+v_0t+\frac{1}{2}gt^2.\]</span> לשלושת האיברים יחידות של מטרים. בדיקת ממדים זו אינה תלויה בכך שערכי הקוד הם מטיפוס <code dir="ltr">Float64</code>.</p>
<p dir="rtl" style="text-align:right">המודל ללא קפיצה אינו מאפשר לכדור להמשיך מתחת לקרקע: <span dir="ltr" class="math display">\[z(t)=\begin{cases}
z_{\mathrm{free}}(t),&amp;z_{\mathrm{free}}(t)&gt;0,\\
0,&amp;z_{\mathrm{free}}(t)\leq0.
\end{cases}\]</span></p>
<p dir="rtl" style="text-align:right">ההגדרה בחלקים מתורגמת ישירות לתנאי:</p>

</div>

```julia
t = 0.35   # s
z0 = 1.20   # m
v0 = 0.0    # m/s
g = -9.81   # m/s^2

if t < 0 || z0 < 0
    throw(ArgumentError("time and initial height must be non-negative"))
end

z_free = z0 + v0 * t + 0.5 * g * t^2   # m

if z_free > 0
    z = z_free
    state = "in air"
else
    z = 0.0
    state = "on the ground"
end

println("at t = $(t) s: z = $(round(z, digits=3)) m ($(state))")
```

<h4 id="כוחות-סימנים-והנחות">כוחות, סימנים והנחות</h4>
<p dir="rtl" style="text-align:right">נבחר <span dir="ltr" class="math inline">\(+x\)</span> במעלה המישור הקבוע ו-<span dir="ltr" class="math inline">\(+y\)</span> ניצב לו. עבור גוף בעל מסה <span dir="ltr" class="math inline">\(m\)</span> (ק"ג), זווית <span dir="ltr" class="math inline">\(\theta\)</span> (רדיאנים ב-<code dir="ltr">sin</code>/<code dir="ltr">cos</code>) וגודל כבידה חיובי <span dir="ltr" class="math inline">\(g\)</span> (m/s<span dir="ltr" class="math inline">\(^2\)</span>), לרכיב הכבידה במורד המישור יש גודל <span dir="ltr" class="math inline">\(mg\sin\theta\)</span> והכוח הנורמלי הוא <span dir="ltr" class="math inline">\(N=mg\cos\theta\)</span>. החיכוך הסטטי יכול להתנגד לתנועה הממשמשת במורד עד <span dir="ltr" class="math inline">\(f_{s,\max}=\mu_sN\)</span>. אם אינו מצליח להחזיק, החיכוך הקינטי <span dir="ltr" class="math inline">\(f_k=\mu_kN\)</span> פועל במעלה המישור ו-<span dir="ltr" class="math inline">\(a=(-mg\sin\theta+\mu_kN)/m\)</span>. אנו מניחים חיכוך קולון, מגע קשיח, ללא גרר, ו-<span dir="ltr" class="math inline">\(0\le\theta&lt;90^\circ\)</span>. המודל אינו מתאר גלגול, מגע משתנה או חיכוך תלוי מהירות. ב-<span dir="ltr" class="math inline">\(\theta=0\)</span> החיכוך הסטטי הנדרש הוא אפס; אם <span dir="ltr" class="math inline">\(\mu_k=0\)</span>, ההחלקה נותנת <span dir="ltr" class="math inline">\(a=-g\sin\theta\)</span>.</p>
<p dir="rtl" style="text-align:right">אנו מאמתים לפני שימוש בקלט. <code dir="ltr">throw(ArgumentError(...))</code> עוצר מיד את החישוב ומודיע לקורא שהתנאי המתועד לקלט לא התקיים. זה שונה מ-<code dir="ltr">println</code>, שמדפיס טקסט בלבד וממשיך. בדיקת השייכות <code dir="ltr">value isa T</code> שואלת אם ערך שייך לטיפוס <code dir="ltr">T</code>; <code dir="ltr">Real</code> הוא שם המשפחה של טיפוסי המספרים הממשיים ב-<code dir="ltr">Julia</code>, הכוללת מספרים שלמים וערכי נקודה צפה רגילים. <code dir="ltr">Bool</code> דורש זהירות מיוחדת כאן: משום ש-<code dir="ltr">true isa Real</code> גם הוא <code dir="ltr">true</code>, בדיקה ל-<code dir="ltr">Real</code> בלבד תקבל <code dir="ltr">true</code> כמסה, ולכן נדחה אותו במפורש.</p>

<p dir="rtl" style="text-align:right"><strong>לפני ההרצה:</strong> נחזה את הערך, הטקסט או הגרף המתוארים לעיל. זו הדוגמה המתאימה מהפרק הכתוב.</p>
</div>

```julia
mass = 2.0      # kg
angle = 30.0    # deg
 μ_s = 0.60
 μ_k = 0.40
g = 9.81        # m/s^2
if !(mass isa Real) || mass isa Bool || !isfinite(mass) || mass <= 0
    throw(ArgumentError("mass must be finite, positive, and non-Boolean"))
elseif !isfinite(angle) || !(0 <= angle < 90) || μ_k < 0 || μ_k > μ_s
    throw(ArgumentError("use 0 <= angle < 90 and 0 <= μ_k <= μ_s"))
end
 θ = deg2rad(angle)   # rad
normal = mass * g * cos(θ)   # N
needed_static = mass * g * sin(θ)   # N
if needed_static <= μ_s * normal
    state = "stuck"
    acceleration = 0.0   # m/s^2
else
    state = "sliding"
    acceleration = (-needed_static + μ_k * normal) / mass   # m/s^2
end
println(state, ", a = ", acceleration, " m/s^2")
```

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>בדיקה:</strong> נשווה את התוצאה להסבר שבהמשך. תא מציג אוטומטית רק את הביטוי האחרון; <code dir="ltr">println</code> מדפיסה טקסט במפורש. לאחר שינוי קלט נריץ שוב את התאים התלויים בו.</p>
</div>

<p dir="rtl" style="text-align:right">פלט הייחוס בפרק הכתוב, להשוואה לפלט שהתקבל בפועל:</p>
<pre dir="ltr" style="text-align:left" class="text"><code dir="ltr">stuck, a = 0.0 m/s^2</code></pre>
</div>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><code dir="ltr">stuck, a = 0.0 m/s^2</code>. כאן <span dir="ltr" class="math inline">\(mg\sin30^\circ=9.81\)</span> N ו-<span dir="ltr" class="math inline">\(\mu_sN\approx10.19\)</span> N, ולכן החיכוך הסטטי מספיק. אם נשנה רק את <code dir="ltr">μ_s</code> ל-<code dir="ltr">0.40</code>, התוצאה תהיה <code dir="ltr">sliding</code> עם תאוצה שלילית: במורד המישור לפי המוסכמה שהצהרנו. שוויון שייך לענף ה-<code dir="ltr">stuck</code>, מוסכמת הסף המפורשת שלנו.</p>
<h2 id="תרגול-משיכה-יכולה-להפוך-את-כיוון-התנועה-הצפוי">תרגול: משיכה יכולה להפוך את כיוון התנועה הצפוי</h2>
<p dir="rtl" style="text-align:right">נוסיף למודל המישור כוח משיכה ידוע <span dir="ltr" class="math inline">\(P\)</span> כלפי מעלה. לפני בחירת סימן החיכוך נחשב את הנטייה ללא חיכוך, <span dir="ltr" class="math inline">\(T=P-mg\sin\theta\)</span>. כאשר <span dir="ltr" class="math inline">\(T\)</span> חיובי, התנועה הצפויה היא במעלה המישור; כאשר הוא שלילי, התנועה הצפויה היא במורד. החיכוך מתנגד לנטייה זו, וחיכוך סטטי יכול להחזיק את הגוף כאשר <span dir="ltr" class="math inline">\(|T|\le\mu_sN\)</span>.</p>
<p dir="rtl" style="text-align:right">נגיע למפגש כשאנו יודעים לקרוא שרשרת תנאים ולהמיר מעלות לרדיאנים. בחלק המודרך הקצר נחזה כיוון, נעקוב אחר ענף אחד ונבדוק את <span dir="ltr" class="math inline">\(T=0\)</span>. לאחר מכן נתחיל את מטלת מעבר הקיר. נתרגל שיטה חוזרת: נאמת קלט, ננסח תנאי פיסיקלי, נחשב את המקרה המתאים ונבדוק את גבולו.</p>
<h2 id="מטלת-החובה-האם-קליע-עובר-מעל-קיר">מטלת החובה: האם קליע עובר מעל קיר?</h2>
<p dir="rtl" style="text-align:right">נשגר מגובה הקרקע במהירות <span dir="ltr" class="math inline">\(v_0\)</span> ובזווית <span dir="ltr" class="math inline">\(\theta\)</span>, עם גודל כבידה <span dir="ltr" class="math inline">\(g&gt;0\)</span>. נגדיר <span dir="ltr" class="math inline">\(+x\)</span> לכיוון הקיר ו-<span dir="ltr" class="math inline">\(+y\)</span> מעלה. במודל של כבידה אחידה ללא גרר, <span dir="ltr" class="math display">\[v_x=v_0\cos\theta,\quad t_w=x_w/v_x,\quad y_w=v_0\sin\theta\,t_w-\tfrac12gt_w^2,\quad v_{y,w}=v_0\sin\theta-gt_w.\]</span> נדרוש <span dir="ltr" class="math inline">\(v_x&gt;0\)</span> לפני החלוקה. נשווה את <span dir="ltr" class="math inline">\(y_w\)</span> לגובה הקיר ונשתמש ב-<span dir="ltr" class="math inline">\(v_{y,w}\)</span> לסיווג עלייה או ירידה. שוויון בגובה נחשב מעבר; מהירות אנכית אפס שייכת לקטגוריית הירידה לפי המוסכמה שפורסמה. גובה התנועה החופשית המחושב עשוי להיות שלילי: עבור קיר שגובהו אינו שלילי, פירושו אי-מעבר ולא המשך פיסיקלי דרך הקרקע.</p>
<p dir="rtl" style="text-align:right">נכתוב תחזית למקרה רגיל ולמקרה גבול, נממש את הקטע הסקלרי ונאמת את התחום המותר. נשווה מחרוזות וערכים בוליאניים בדיוק, וגבהים בקירוב לפי הסבולת שפורסמה. נסביר את ההבדל בין החלטת המעבר לסיווג התנועה: אפשר לעבור את הקיר גם בירידה. דף המטלה הוא המקום שבו מפורטים שמות הקלט, הבדיקות, מוסכמת הפונקציה העוטפת וההגשה; אין צורך בלולאות או בהגדרת פונקציות.</p>
<h2 id="בונוס-לבחירה-יחידות-ופאזורים-בזרם-חילופין">בונוס לבחירה: יחידות ופאזורים בזרם חילופין</h2>
<p dir="rtl" style="text-align:right">הבונוס הוא רשות, אינו תנאי להמשך ואינו נבחן. הוא מזכה ב<strong>שליש נקודה בציון הסופי</strong>.</p>
<p dir="rtl" style="text-align:right">הקריאה העצמאית מפתחת מספרים מרוכבים, גודל, פאזה, צמוד וקשר אוילר לפני השימוש ב-<code dir="ltr">cis</code>. נבחר משרעות שיא ואת מוסכמת הזמן <span dir="ltr" class="math inline">\(\exp(+i\omega t)\)</span>, נקבע את פאזת המתח לאפס, נבנה <span dir="ltr" class="math inline">\(Z=|Z|\exp(i\phi_Z)\)</span> ונחשב <span dir="ltr" class="math inline">\(I=V/Z\)</span>. הזרם המדיד הוא החלק הממשי של <span dir="ltr" class="math inline">\(I\exp(i\omega t)\)</span>; המספר המרוכב הוא ייצוג ולא זרם נמדד נוסף.</p>
<p dir="rtl" style="text-align:right">נעקוב אחר דוגמאות <code dir="ltr">Unitful</code>, נוודא שלמתח חלקי התנגדות יש יחידות זרם ונבדוק נגד בעל פאזה אפס לפני מקרה של רבע מחזור. נגיש קטע המציב זרם מרוכב באמפר ונריץ את בדיקות היחידות השקולות והקלט הפסול. נסביר מדוע פאזה חיובית של העכבה נותנת פאזה שלילית של הזרם. המודל לינארי, בעל תדירות אחת ובמצב מתמיד; מעברי מיתוג ורכיבים לא-לינאריים אינם מתוארים בו.</p>
<h2 id="להעמקה-החזרה-פנימית-מלאה">להעמקה: החזרה פנימית מלאה</h2>
<p dir="rtl" style="text-align:right">נשתמש במתכון המסופק לחוק סנל <span dir="ltr" class="math inline">\(n_1\sin\theta_1=n_2\sin\theta_2\)</span>, עם מקדמי שבירה חיוביים וזוויות הנמדדות מהנורמל לממשק. עבור <span dir="ltr" class="math inline">\(n_1&gt;n_2\)</span>, זווית פגיעה הגדולה מהזווית הקריטית <span dir="ltr" class="math inline">\(\theta_c=\arcsin(n_2/n_1)\)</span> גורמת להחזרה פנימית מלאה. נסווג דוגמאות מתחת לסף, על הסף ומעליו לפני חישוב סינוס הפוך; הפעלת <code dir="ltr">asin</code> על מספר גדול מאחת ללא בדיקה אינה חישוב פיסיקלי תקין.</p>
<p dir="rtl" style="text-align:right">נשרטט החזרות אידיאליות חוזרות בסיב ישר ונשנה את זווית הכניסה. נרשום אילו בחירות נשארות מונחות בסיב ונסביר מהו הנורמל בכל ממשק. לולאת עקיבה מסופקת, אם נשתמש בה, היא הצצה לבחירה ולא חומר חובה בשבוע 2. אופני גל, נפיצה, הפסדי כיפוף ובליעה דורשים פיסיקה נוספת. החקירה נפרדת מבונוס הפאזורים, אינה נבחנת ואינה תנאי להמשך.</p>
