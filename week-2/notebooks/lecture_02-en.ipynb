## Piecewise Kinematics as a Decision

#### Signed acceleration versus positive gravity magnitude

Height is positive upward, so gravitational acceleration is $$g=-9.81\ \mathrm{m/s^2}.$$ For initial height $z_0$ and velocity $v_0$, $$z_{\mathrm{free}}(t)=z_0+v_0t+\frac{1}{2}gt^2.$$ The three terms have units of metres. This dimensional check is independent of whether their code values are `Float64`.

Our no-bounce model does not allow the ball to continue below the ground: $$z(t)=\begin{cases}
z_{\mathrm{free}}(t),&z_{\mathrm{free}}(t)>0,\\
0,&z_{\mathrm{free}}(t)\leq0.
\end{cases}$$

The piecewise definition translates directly to a conditional:

**Before running:** predict the value, printed text, or graph described above. This is the corresponding example from the written chapter.

````julia
t = 0.35   # s
z0 = 1.20   # m
v0 = 0.0    # m/s
g = -9.81   # m/s^2

if t < 0 || z0 < 0
    throw(ArgumentError("time and initial height must be non-negative"))
end

z_free = z0 + v0 * t + 0.5 * g * t^2   # m

if z_free > 0
    z = z_free
    state = "in air"
else
    z = 0.0
    state = "on the ground"
end

println("at t = $(t) s: z = $(round(z, digits=3)) m ($(state))")
````

## A piecewise friction model on an incline

#### Forces, signs, and assumptions

Take $+x$ uphill along a fixed incline and $+y$ normal to it. For a block of mass $m$ (kg), angle $\theta$ (radians in `sin`/`cos`), and positive gravity magnitude $g$ (m/s$^2$), gravity has downhill component $mg\sin\theta$ and normal force $N=mg\cos\theta$. Static friction can oppose the impending downhill motion up to $f_{s,\max}=\mu_sN$. If it cannot hold, kinetic friction $f_k=\mu_kN$ acts uphill and $a=(-mg\sin\theta+\mu_kN)/m$. We assume Coulomb friction, rigid contact, no drag, and $0\le\theta<90^\circ$. The model does not describe rolling, changing contact, or velocity-dependent friction. At $\theta=0$, the required static friction is zero; if $\mu_k=0$, sliding gives $a=-g\sin\theta$.

We validate before using an input. `throw(ArgumentError(...))` immediately stops the calculation and tells its caller that a documented input condition failed. It is different from `println`, which merely reports text and continues. The membership test `value isa T` asks whether a value belongs to type `T`; `Real` names Julia's family of real-number types, which includes ordinary integers and floating-point values. `Bool` needs special care here: because `true isa Real` is also `true`, a check for `Real` alone would accept `true` as a mass, so we reject it explicitly.

**Before running:** predict the value, printed text, or graph described above. This is the corresponding example from the written chapter.

````julia
mass = 2.0      # kg
angle = 30.0    # deg
 μ_s = 0.60
 μ_k = 0.40
g = 9.81        # m/s^2
if !(mass isa Real) || mass isa Bool || !isfinite(mass) || mass <= 0
    throw(ArgumentError("mass must be finite, positive, and non-Boolean"))
elseif !isfinite(angle) || !(0 <= angle < 90) || μ_k < 0 || μ_k > μ_s
    throw(ArgumentError("use 0 <= angle < 90 and 0 <= μ_k <= μ_s"))
end
 θ = deg2rad(angle)   # rad
normal = mass * g * cos(θ)   # N
needed_static = mass * g * sin(θ)   # N
if needed_static <= μ_s * normal
    state = "stuck"
    acceleration = 0.0   # m/s^2
else
    state = "sliding"
    acceleration = (-needed_static + μ_k * normal) / mass   # m/s^2
end
println(state, ", a = ", acceleration, " m/s^2")
````

## Recitation: a pull can reverse impending motion

We add a known uphill pull $P$ to the incline model. Before choosing a friction sign, we calculate the friction-free tendency $T=P-mg\sin\theta$. Positive $T$ means impending uphill motion; negative $T$ means impending downhill motion. Friction opposes that tendency, and static friction can hold whenever $|T|\le\mu_sN$.

## Core homework: does a projectile clear a wall?

We launch from ground level with speed $v_0$, angle $\theta$, and gravity magnitude $g>0$, with $+x$ toward the wall and $+y$ upward. In the uniform-gravity, no-drag model, $$v_x=v_0\cos\theta,\quad t_w=x_w/v_x,\quad y_w=v_0\sin\theta\,t_w-\tfrac12gt_w^2,\quad v_{y,w}=v_0\sin\theta-gt_w.$$ We require $v_x>0$ before dividing. We compare $y_w$ with the wall height, and use $v_{y,w}$ to classify rising or falling. Equality in height counts as clearing; zero vertical velocity belongs to the published falling category. The computed free-flight height may be negative: for a nonnegative wall height this means failure to clear, not motion physically continuing through the ground.

We choose peak amplitudes and the time convention $\exp(+i\omega t)$, set the voltage phase to zero, build $Z=|Z|\exp(i\phi_Z)$, and calculate $I=V/Z$. The measurable current is the real part of $I\exp(i\omega t)$; the complex number is a representation, not an additional measured current.

We follow the supplied Unitful examples, verify that voltage divided by resistance has current units, and check a resistor with zero phase before trying a quarter-cycle phase. We explain why a positive impedance phase gives a negative current phase. This is a single-frequency linear steady-state model; switching transients and nonlinear components are outside its scope.

## Going further: total internal reflection

We use the supplied Snell-law recipe $n_1\sin\theta_1=n_2\sin\theta_2$, with positive refractive indices and angles measured from the interface normal. For $n_1>n_2$, incidence beyond $\theta_c=\arcsin(n_2/n_1)$ gives total internal reflection. We classify below-critical, critical, and above-critical examples before evaluating an inverse sine; blindly taking `asin` of a number greater than one is not a physical calculation.

We sketch repeated ideal reflections in a straight fiber and vary the launch angle. We record which choices remain guided and explain the normal used at each interface. Any supplied tracing loop is an optional preview, not Week 2 core work. Waveguide modes, dispersion, bending losses, and absorption require more physics than this ray model.
