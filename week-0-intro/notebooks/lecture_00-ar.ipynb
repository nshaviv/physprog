<h2 id="ما-هو-البرنامج-وما-هو-الحاسوب">ما هو البرنامج؟ وما هو الحاسوب؟</h2>
<p dir="rtl" style="text-align:right">الـ<em>برنامج</em> تسلسل من التعليمات يخبر الحاسوب كيف ينفّذ مهمة. ففي مثال المقذوف، يمكن أن يبدأ بموضع وسرعة، ويحسب التسارع، ويحدّث الحركة عند أزمنة متتالية. ولكتابة تلك التعليمات، نحتاج إلى تحديد البيانات التي نخزّنها، وكيف نحسب بها، ومتى نكرّر أو نتوقّف. وغالبًا ما يستطيع القارئ البشري أن يكمل خطوة ناقصة في اشتقاق؛ أما الحاسوب فيحتاج إلى تلك الخطوة مكتوبة صراحةً.</p>
<p dir="rtl" style="text-align:right">الـ<em>حاسوب</em> آلة فيزيائية تنفّذ هذه التعليمات مرارًا. فمعالجه المركزي (<code dir="ltr">CPU</code>) ينفّذ العمليات الحسابية والمنطقية؛ والذاكرة (<code dir="ltr">RAM</code>) تحمل البيانات والتعليمات قيد الاستخدام؛ والتخزين الدائم يحفظ الملفات بعد انتهاء البرنامج؛ وأجهزة الإدخال/الإخراج تصل الآلة بالملفات والشاشات والأجهزة والشبكات. وعندما نشغّل سكربت <code dir="ltr">Julia</code>، يترجم نظام لغة <code dir="ltr">Julia</code> تعليماتنا إلى عمليات يستطيع المعالج تنفيذها، بينما يدير نظام التشغيل الملفات والذاكرة والوصول إلى الشاشة.</p>
<p dir="rtl" style="text-align:right">بينما نقرأ برنامجًا، يمكننا تتبّع البيانات من القيم الابتدائية عبر الحسابات وصولًا إلى النتيجة. وتصف <em>الخوارزمية</em> خطوات ذلك الحساب؛ ويساعدنا النموذج الفيزيائي على تفسير ما ينتج. وسنبني طريقة العمل هذه تدريجيًا. فيقدّم الأسبوع 2 الأنواع وتحويل المدخلات والقرارات؛ ويقدّم الأسبوع 3 الحلقات وقواعد التوقّف؛ ويقدّم الأسبوع 5 تعريفات الدوال.</p>
<h2 id="لماذا-نبدأ-بـjulia">لماذا نبدأ بـ<code dir="ltr">Julia</code>؟</h2>
<p dir="rtl" style="text-align:right">صُمّمت <code dir="ltr">Julia</code> للعمل العددي والعلمي. لديها تدوين مباشر للمتجهات والمصفوفات والجبر الخطي وتعددية الإرسال وحلّالات المعادلات التفاضلية، مع بقائها مناسبة للسكربتات والدفاتر الاستكشافية. وهي سريعة بما يكفي بحيث تكون النسخة الأولى القابلة للقراءة من نموذج عددي مفيدة غالبًا بالفعل.</p>
<p dir="rtl" style="text-align:right">بالنسبة لنا كفيزيائيين، الفائدة الأولى المهمة هي أن الكود يمكن أن يبدو مثل النموذج. فمتجه القياسات، والمصفوفة التي تمثّل تحويلًا خطيًا، ومصفوفة قيمٍ مُعيّنة زمنيًا، كلها كائنات عادية في <code dir="ltr">Julia</code>. والتمييز بين قيمة مفردة ومتجه وحسابٍ عنصرًا بعنصر واضح في التدوين. وهذا يجعل الأبعاد والفهارس والوحدات والعملية الفيزيائية المقصودة أسهل في الفحص قبل أن يُدفن خطأ عددي داخل برنامج طويل.</p>
<p dir="rtl" style="text-align:right">تتّصف <code dir="ltr">Julia</code> بأنها معبّرة لأن تدوينها يتيح لنا كتابة العملية قريبة من الرياضيات، وقابلة للتركيب لأن العمليات المحدّدة جيدًا وأدوات الحزم يمكن دمجها في حساب أكبر. فمثلًا يمكن أن تدخل عملية على مصفوفة في حساب آخر، بينما يمكن تعريف الفكرة العددية نفسها للأعداد والمتجهات والمصفوفات، أو لاحقًا للقيم التي تحمل وحدات. وتختار <code dir="ltr">Julia</code> الطريقة المناسبة وفق أنواع المدخلات؛ وتُسمّى هذه الفكرة <em>تعددية الإرسال</em>. سنستخدم الأدوات العلمية أولًا، ثم نعود إلى هذه الفكرة التصميمية عندما تساعدنا على شرح كيف تعمل عملية واحدة في عدة سياقات فيزيائية.</p>
<p dir="rtl" style="text-align:right">يمكن لمشاريع <code dir="ltr">Julia</code> أن تسجّل الحزم والنسخ المستخدمة في تحليلٍ ما في ملفات المشروع. وهذا يجعل من العملي العودة إلى حسابٍ بعد أشهر أو مشاركته مع زميل دون تغيير صامت لأدوات الحساب التي يقوم عليها. وللغة أيضًا مكتبة قياسية قوية للجبر الخطي العددي ومنظوم بيئي متنامٍ للمعادلات التفاضلية والاستمثال والإحصاء والوحدات والرسم وتعلّم الآلة والبيانات العلمية. سنبني الطرق الأساسية بأنفسنا أولًا، ثم نستخدم تلك الحزم كتطبيقات مُفحَصة بعناية لأفكار فهمناها بالفعل.</p>
<p dir="rtl" style="text-align:right">السرعة مهمة في الفيزياء لأنها تتيح لنا اختبار نموذج، وفحص نتيجة، وتغيير افتراض، والتكرار. ويلخّص الشكل أدناه مجموعة الاختبارات المصغّرة لمشروع <code dir="ltr">Julia</code>. يقارن محوره الأفقي زمن التنفيذ بلغة <code dir="ltr">C</code> على مقياس لوغاريتمي، فالنقاط الأبعد إلى اليسار أسرع؛ أما محوره الرأسي فهو طول تطبيق الاختبار. وتقع <code dir="ltr">Julia</code> قرب اللغات المترجَمة في هذه الأنوية العددية الصغيرة بينما يبقى كود مثالها مدمجًا. وتصف هذه التوقيتات الأنوية المختارة؛ أما سرعة برنامج كامل فتعتمد أيضًا على خوارزميته وتوزيع بياناته وتخصيصاته وطريقة قياسنا له. والبيانات والبرامج متاحة في <a href="https://github.com/JuliaLang/Microbenchmarks">مستودع اختبارات <code dir="ltr">Julia</code> المصغّرة</a>.</p>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right">هنا <code dir="ltr">x</code> و<code dir="ltr">u</code> و<code dir="ltr">y</code> متجهات، بينما <code dir="ltr">A</code> و<code dir="ltr">B</code> و<code dir="ltr">C</code> مصفوفات. يمكننا مقارنة كل سطر من الكود بمعادلته مباشرة. وبينما نفعل ذلك، يساعدنا فحص أبعاد المصفوفات والوحدات الفيزيائية على رؤية ما إذا كان الحساب يعبّر عن النموذج الذي قصدناه.</p>
<p dir="rtl" style="text-align:right">وهناك تفاصيل عملية نعتاد عليها. فقد يستغرق التشغيل الأول لكود <code dir="ltr">Julia</code> الجديد وقتًا أطول بينما تترجم <code dir="ltr">Julia</code> كود آلة متخصصًا؛ وقد تكون التشغيلات التالية أسرع بكثير. ومجتمعها أصغر من مجتمع <code dir="ltr">Python</code>، لذا قد تؤثر الحزم المتاحة في اختيارنا للأداة. وسنتعلّم قياس الأداء، وسنستخدم <code dir="ltr">Python</code> عندما يناسب منظومها البيئي المهمة.</p>
<p dir="rtl" style="text-align:right">تدعم <code dir="ltr">Julia</code> أيضًا إدخال رموز يونيكود بأسلوب <code dir="ltr">LaTeX</code> في حلقة القراءة–التقييم–الطباعة (<code dir="ltr">REPL</code>) وفي <code dir="ltr">VS Code</code>. ولكتابة تعبير قياسي مثل <span dir="ltr" class="math display">\[v_1^{\prime}\cdot(A v_2),\]</span> نكتب أمرًا بشرطة مائلة للخلف ونضغط <code dir="ltr">Tab</code> بعد كل رمز:</p>
<ul>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">\alpha&lt;Tab&gt;</code> ينتج <span dir="ltr" class="math inline">\(\alpha\)</span>.</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">\cdot&lt;Tab&gt;</code> ينتج <span dir="ltr" class="math inline">\(\cdot\)</span>.</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">v\_1&lt;Tab&gt;\prime&lt;Tab&gt; \cdot&lt;Tab&gt; (A * v\_2&lt;Tab&gt;)</code> ينتج <span dir="ltr" class="math inline">\(\mathrm{v}_{1}^{\prime}\mathbin{\cdot}(\mathrm{A} * \mathrm{v}_{2})\)</span>.</p></li>
</ul>

<p dir="rtl" style="text-align:right">اندمج ثقبان أسودان وأصدرا موجات جاذبية. وتغيّر موجة عابرة الطول النسبي لذراعَي مقياس التداخل: فالانفعال <span dir="ltr" class="math inline">\(h=\Delta L/L\)</span> كمية بلا أبعاد. وتحتوي مخرجات الكاشف على ضوضاء إلى جانب أي إشارة. ويكتم مرشح النطاق الترددات خارج 50–250 هرتز، وتكتم مرشحات الثلم توافقات ضيقة لشبكة الكهرباء، ويتجنّب الترشيح إلى الأمام والخلف تأخيرًا طوريًا مصطنعًا. والترشيح وحده لا يثبت وجود كشف. أما توافق كاشفين مستقلين بعد التأخير والتصحيح الاتجاهي الموثّقين فهو دليل أقوى بكثير من أثر واحد، لكن التحليل الحقيقي يقدّم أيضًا دلالة الضوضاء إحصائيًا ويقارن نماذج شكل الموجة.</p>

<p dir="rtl" style="text-align:right">يقدّم الفصل الثاني <code dir="ltr">Python</code> بنقل طرق فهمناها أصلًا، ثم يوسّعها إلى مسائل القيم الذاتية، ومحاكاة مونت كارلو، والملاءمة غير الخطية والملاءمة الواعية بعدم اليقين، وتحليل فورييه، والمعادلات التفاضلية الجزئية، والأداء، وتعلّم الآلة، وسير العمل الرمزية. وتوفّر واجبات الأسبوع تدريبًا لازمًا وفحوص جاهزية؛ ويعطي التقييم المحوسب وزنًا أكبر لحلّ مسألة غير مألوفة باستقلالية وشرح المعنى الفيزيائي للنتيجة.</p>
<h2 id="ما-الذي-سنتعلمه">ما الذي سنتعلمه؟</h2>
<p dir="rtl" style="text-align:right">لتحويل مسألة فيزيائية إلى برنامج، نحتاج أولًا إلى تحديد كيف سيسير الحساب. بأي قيم نبدأ؟ وما الذي يتغيّر في كل خطوة، ومتى ينبغي أن نتوقّف؟ سنتعلّم التعبير عن هذه الخيارات من خلال <em>التفكير الخوارزمي</em>، باختيار تمثيل للبيانات وتقسيم الحساب إلى خطوات نستطيع متابعتها واختبارها. وستساعدنا الحالات البسيطة والتوقعات الفيزيائية على فهم المخرجات.</p>
<p dir="rtl" style="text-align:right">ثم سنستخدم تلك العادات من أجل:</p>
<ul>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">قراءة بيانات تجريبية ورصدية من الملفات؛</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">تصوير البيانات والنماذج والبواقي وعدم اليقين؛</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">ملاءمة نماذج خطية وغير خطية وتفسير معاملاتها؛</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">حلّ المعادلات وتكامل الدوال وحلّ المعادلات التفاضلية العادية عدديًا؛</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">العمل مع المتجهات والمصفوفات وتحويلات فورييه والعمليات العشوائية والمعادلات التفاضلية الجزئية البسيطة؛</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">كتابة كود علمي مقروء وقابل للاختبار وقابل لإعادة الإنتاج بـ<code dir="ltr">Julia</code> و<code dir="ltr">Python</code>.</p></li>

<h2 id="تشويق-إعادة-تحليل-كبلر-للمريخ">تشويق: إعادة تحليل كبلر للمريخ</h2>
<p dir="rtl" style="text-align:right">في نهاية الفصل الأول سنستخدم أرصادًا مرتبطة ببرنامج تيكو براهي لرصد المريخ لإعادة بناء نسخة من حجة كبلر. وستصبح مجموعة البيانات هذه دليلًا لاختبار قانون كبلر الأول: مدارات الكواكب إهليلجات والشمس في إحدى بؤرتيها. سنأخذ أرصادًا جُمعت في تواريخ مختلفة، ونحوّل المواضع الزاوية المجدولة إلى وصف هندسي مشترك، ونلائم نماذج مدارية متنافسة، ونبيّن كميًا لماذا ينجح الإهليلج حيث تفشل الدائرة.</p>
