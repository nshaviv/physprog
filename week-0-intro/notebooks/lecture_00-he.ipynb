<h2 id="למה-להתחיל-ב-julia">למה להתחיל ב-<code dir="ltr">Julia</code>?</h2>
<p dir="rtl" style="text-align:right"><code dir="ltr">Julia</code> נבנתה לעבודה נומרית ומדעית. יש לה תחביר ישיר עבור וקטורים, מטריצות, אלגברה לינארית, שיגור מרובה ופותרי משוואות דיפרנציאליות, והיא מתאימה גם לסקריפטים ולמחברות חקירה. היא גם מהירה מספיק כך שלעיתים קרובות הגרסה הקריאה הראשונה של מודל נומרי כבר שימושית.</p>
<p dir="rtl" style="text-align:right">עבורנו כפיסיקאים, היתרון הראשון והחשוב הוא שהקוד יכול להיראות כמו המודל. וקטור של מדידות, מטריצה שמייצגת העתקה לינארית ומערך של ערכים שנדגמו בזמן הם אובייקטים רגילים ב-<code dir="ltr">Julia</code>. ההבחנה בין ערך יחיד, וקטור וחישוב איבר-איבר נראית בתחביר. כך קל יותר לבדוק ממדים, אינדקסים, יחידות והפעולה הפיסיקלית שהתכוונו אליה, לפני שטעות נומרית נקברת בתוכנית ארוכה.</p>
<p dir="rtl" style="text-align:right"><code dir="ltr">Julia</code> מאפשרת לנו לכתוב פעולה קרוב לצורה המתמטית שלה, ולחבר פעולות וכלים מוגדרים היטב לחישוב גדול יותר. למשל, אפשר להעביר תוצאה של פעולת מטריצה לחישוב הבא; ואותו רעיון נומרי יכול לפעול על מספרים, וקטורים, מטריצות ובהמשך גם על ערכים שנושאים יחידות. <code dir="ltr">Julia</code> בוחרת את המימוש המתאים לפי טיפוסי הקלט; הרעיון הזה נקרא <em>multiple dispatch</em>. בתחילת הקורס נשתמש בכלים המדעיים, ובהמשך נחזור לרעיון התכנוני הזה כשיעזור להבין כיצד פעולה אחת משתלבת בכמה הקשרים פיסיקליים.</p>
<p dir="rtl" style="text-align:right">פרויקטים ב-<code dir="ltr">Julia</code> יכולים לרשום בקובצי פרויקט את החבילות ואת הגרסאות ששימשו לניתוח. כך אפשר לחזור לחישוב כעבור חודשים או לשתף אותו עם משתף פעולה, בלי להחליף בשקט את כלי החישוב שמתחתיו. לשפה יש גם ספרייה בסיסית חזקה לאלגברה לינארית נומרית ואקוסיסטם מתרחב למשוואות דיפרנציאליות, אופטימיזציה, סטטיסטיקה, יחידות, גרפים, למידת מכונה ונתונים מדעיים. תחילה נבנה בעצמנו את השיטות המרכזיות, ובהמשך נשתמש בחבילות כמימושים שנבדקו היטב של רעיונות שכבר הבנו.</p>
<p dir="rtl" style="text-align:right">מהירות חשובה בפיסיקה משום שהיא מאפשרת לבדוק מודל, לבחון תוצאה, לשנות הנחה ולחזור על החישוב. האיור הבא מסכם את חבילת מבחני הביצועים הזעירים של פרויקט <code dir="ltr">Julia</code>. הציר האופקי משווה את זמן הריצה ל-<code dir="ltr">C</code> בסקלה לוגריתמית, ולכן נקודות שמאלה יותר מהירות יותר; הציר האנכי הוא אורך המימוש של המבחן. במבחנים נומריים קטנים אלה <code dir="ltr">Julia</code> קרובה לשפות מהודרות, בעוד קוד הדוגמה נשאר תמציתי. תזמונים אלה מתארים את המבחנים הנבחרים בלבד; מהירותה של תוכנית שלמה תלויה גם באלגוריתם שלה, בארגון הנתונים, בהקצאת הזיכרון ובאופן שבו אנחנו מודדים אותה. הנתונים והתוכניות זמינים ב-<a href="https://github.com/JuliaLang/Microbenchmarks" class="uri">https://github.com/JuliaLang/Microbenchmarks</a>.</p>

<h3 id="ערכים-מסופקים-להדגמת-הכתיב">ערכים מסופקים להדגמת הכתיב</h3>
<p dir="rtl" style="text-align:right">בדוגמה הכתובה מופיעות מטריצות ווקטורים סמליים. נספק ערכים קטנים וחסרי ממד כדי שאפשר יהיה להריץ אותה. עבודה במטריצות שייכת להמשך; זו המחשה שאינה נבחנת.</p>

```julia
A = [1.0 0.1; 0.0 1.0]
B = [0.0; 0.1;;]
C = [1.0 0.0]
x = [2.0, 3.0]
u = [4.0];
```

```julia
x_next = A * x + B * u
y = C * x
```

<p dir="rtl" style="text-align:right">כאן <code dir="ltr">x</code>, <code dir="ltr">u</code> ו-<code dir="ltr">y</code> הם וקטורים, ואילו <code dir="ltr">A</code>, <code dir="ltr">B</code> ו-<code dir="ltr">C</code> הן מטריצות. נוכל להשוות כל שורת קוד ישירות למשוואה שלה. בדיקת ממדי המטריצות והיחידות הפיסיקליות תעזור לנו לראות אם החישוב מבטא את המודל שהתכוונו אליו.</p>
<p dir="rtl" style="text-align:right">יש גם פרטים מעשיים שנכיר במהלך העבודה. ההרצה הראשונה של קוד חדש ב-<code dir="ltr">Julia</code> עלולה להימשך יותר זמן בזמן שהשפה מהדרת קוד מכונה מותאם; הרצות נוספות עשויות להיות מהירות הרבה יותר. הקהילה שלה קטנה מזו של <code dir="ltr">Python</code>, ולכן החבילות הזמינות יכולות להשפיע על בחירת הכלי. נלמד למדוד ביצועים ונשתמש ב-<code dir="ltr">Python</code> כאשר האקוסיסטם שלה מתאים למשימה.</p>
<p dir="rtl" style="text-align:right">ב-<code dir="ltr">Julia</code> אפשר להקליד ב-<code dir="ltr">REPL</code> וב-<code dir="ltr">VS Code</code> סימנים מתמטיים בעזרת פקודות דמויות <code dir="ltr">LaTeX</code>. למשל, כדי לכתוב ביטוי סקלרי כגון <span dir="ltr" class="math display">\[v_1^{\prime}\cdot(A v_2),\]</span> מקלידים פקודה שמתחילה בלוכסן הפוך ולוחצים <code dir="ltr">Tab</code> אחרי כל סימן:</p>
<ul>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">\alpha&lt;Tab&gt;</code> יוצר <span dir="ltr" class="math inline">\(\alpha\)</span>.</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">\cdot&lt;Tab&gt;</code> יוצר <span dir="ltr" class="math inline">\(\cdot\)</span>.</p></li>
<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right"><code dir="ltr">v\_1&lt;Tab&gt;\prime&lt;Tab&gt; \cdot&lt;Tab&gt; (A * v\_2&lt;Tab&gt;)</code> יוצר <span dir="ltr" class="math inline">\(\mathrm{v}_{1}^{\prime}\mathbin{\cdot}(\mathrm{A} * \mathrm{v}_{2})\)</span>.</p></li>
</ul>
<p dir="rtl" style="text-align:right">מתקבלת הצורה המתמטית שמופיעה למעלה: בשמות משתנים אפשר להשתמש באותיות יווניות, באינדקסים תחתונים, בגרש ובאופרטורים שמופיעים בחישוב.</p>

<p dir="rtl" style="text-align:right">חלק גדול מן הקוד הנומרי ב-<code dir="ltr">Python</code> מעביר את החישובים הכבדים לספריות מהודרות, כמו <code dir="ltr">NumPy</code> ו-<code dir="ltr">SciPy</code>. כך ביטוי קצר ב-<code dir="ltr">Python</code> יכול לבצע עבודה נומרית רבה ביעילות. לכן זמן הריצה של לולאה שנכתבה ב-<code dir="ltr">Python</code> רגילה עשוי להיות שונה מאוד מזה של חישוב שמבצעת ספרייה נומרית. כשנבחן ביצועים, נבדוק היכן העבודה מתבצעת ונמדוד כמה זמן היא נמשכת.</p>
