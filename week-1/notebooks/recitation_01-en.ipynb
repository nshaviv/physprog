#### Student physics box: a spring

For a spring stretched a modest distance in its elastic regime, the restoring force magnitude is $F=kx$. Here extension $x$ is measured from the unloaded length in m, force $F$ is the applied force magnitude in N, and slope $k$ is the spring constant in N/m. We use positive extension and positive applied-force magnitude. At $x=0$, an ideal spring predicts $F=0$. Preload, an offset, hysteresis, plastic deformation, and leaving the elastic regime can invalidate a straight-line description.

### 5--13 minutes: one guided fit
The supplied notebook uses the same centered-sum recipe as the lecture. We read its result rather than redesign it.

**Before running:** predict the value, printed text, or graph described above. This is the corresponding example from the written chapter.

````julia
x = [0.00, 0.02, 0.04, 0.06, 0.08, 0.10]   # m
y = [0.01, 0.39, 0.81, 1.18, 1.61, 1.99]    # N
n = length(x)
xbar = sum(x) / n; ybar = sum(y) / n
slope = sum((x .- xbar) .* (y .- ybar)) / sum((x .- xbar).^2)
intercept = ybar - slope * xbar
println("slope = $(round(slope; digits=1)) N/m")
println("intercept = $(round(intercept; digits=4)) N")
````

## Going further: Lissajous curves

In this optional, unexamined investigation we use a supplied plotting recipe for $$x(t)=A\cos(at+\delta),\qquad y(t)=B\sin(bt).$$ Here $t$ is in seconds, $a,b$ are angular frequencies in rad/s, $\delta$ is a phase in radians, and $A,B$ set the coordinate amplitudes. These curves describe two independent sinusoidal motions viewed together; they are not trajectories under a central gravitational force.

We keep amplitudes fixed and compare frequency ratios $1:1$, $2:1$, and $3:2$. Then, for one ratio, we vary the phase from $0$ to $\pi/2$. We save two contrasting plots and write a conjecture about closure and symmetry. We check whether a longer time interval changes an apparently open curve: a finite plotting window is not evidence of nonclosure.
