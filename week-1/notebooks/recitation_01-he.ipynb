<h3 id="דקות-התאמה-מודרכת-אחת">5–13 דקות: התאמה מודרכת אחת</h3>
<p dir="rtl" style="text-align:right">המחברת משתמשת במתכון הסכומים הממורכזים של ההרצאה. אנו קוראים את תוצאתו ולא מתכננים אותו מחדש.</p>

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>לפני ההרצה:</strong> נחזה את הערך, הטקסט או הגרף המתוארים לעיל. זו הדוגמה המתאימה מהפרק הכתוב.</p>
</div>

```julia
x = [0.00, 0.02, 0.04, 0.06, 0.08, 0.10]   # m
y = [0.01, 0.39, 0.81, 1.18, 1.61, 1.99]    # N
n = length(x)
xbar = sum(x) / n; ybar = sum(y) / n
slope = sum((x .- xbar) .* (y .- ybar)) / sum((x .- xbar).^2)
intercept = ybar - slope * xbar
println("slope = $(round(slope; digits=1)) N/m")
println("intercept = $(round(intercept; digits=4)) N")
```

<div dir="rtl" style="text-align:right">
<p dir="rtl" style="text-align:right"><strong>בדיקה:</strong> נשווה את התוצאה להסבר שבהמשך. תא מציג אוטומטית רק את הביטוי האחרון; <code dir="ltr">println</code> מדפיסה טקסט במפורש. לאחר שינוי קלט נריץ שוב את התאים התלויים בו.</p>
</div>

<div dir="rtl" style="text-align:right">

<pre dir="ltr" style="text-align:left" class="text"><code dir="ltr">slope = 19.9 N/m
intercept = 0.0033 N</code></pre>
</div>

<h2 id="תרגול-מדידת-קבוע-קפיץ">תרגול: מדידת קבוע קפיץ</h2>
<p dir="rtl" style="text-align:right">נעביר את אותו ניתוח לקפיץ. זוגות המדידות הם התארכות <span dir="ltr" class="math inline">\(x\)</span> במטרים וכוח מופעל <span dir="ltr" class="math inline">\(F\)</span> בניוטון; נתאים את המודל <span dir="ltr" class="math inline">\(F=kx+F_0\)</span>. כאן, בניגוד לגרף שבו ההתארכות בציר האנכי, השיפוע הוא ישירות קבוע הקפיץ <span dir="ltr" class="math inline">\(k\)</span> ביחידות ניוטון למטר. נצפה מראש לשיפוע חיובי ונבדוק שלחותך יש יחידות כוח.</p>

<h2 id="מטלת-החובה-הסקת-תאוצת-הכובד-ממדידות-נפילה-חופשית">מטלת החובה: הסקת תאוצת הכובד ממדידות נפילה חופשית</h2>

<li dir="rtl" style="text-align:right"><p dir="rtl" style="text-align:right">נחזה את צורת הגרפים של <span dir="ltr" class="math inline">\(y\)</span> מול <span dir="ltr" class="math inline">\(t\)</span> ושל <span dir="ltr" class="math inline">\(y\)</span> מול <span dir="ltr" class="math inline">\(t^2\)</span> לפני ההתאמה. ניישם את המתכון ונשמור את התחזיות ואת השאריות, מדידה פחות תחזית.</p></li>

<h2 id="בונוס-לבחירה-גרף-שמעביר-תוצאה-מדעית">בונוס לבחירה: גרף שמעביר תוצאה מדעית</h2>
<p dir="rtl" style="text-align:right">הבונוס הוא רשות, אינו תנאי להמשך ואינו נבחן. השלמתו מזכה ב<strong>שליש נקודה בציון הסופי</strong>.</p>
<p dir="rtl" style="text-align:right">לאחר קריאה עצמאית של 10–15 דקות נשפר גרף מינימלי מסופק של חוק אוהם. כאן המתח בציר האופקי והזרם בציר האנכי, ולכן שיפוע המודל הוא מוליכות בסימנס ולא התנגדות באוהם. נזהה את סדרת הזרם הנמדד ואת סדרת הזרם המותאם לפני שינוי העיצוב.</p>
<p dir="rtl" style="text-align:right">נוסיף את הכותרת, תוויות הצירים והיחידות, המקרא, סמני המדידות, קו המודל, המסגרת וגודל האיור שנקבעו, ונייצא <code dir="ltr">PNG</code> ו-<code dir="ltr">PDF</code>. קטע החישוב מספק סדרות מאומתות ותואמות; התמונות נבדקות בנפרד לפי הרשימה המפורשת. נוודא ששינוי תווית או סגנון קו אינו משנה מדידה. דף הבונוס מציין את הטקסטים ושמות הקבצים המדויקים, כך שהבדיקה עוסקת בשלמות מדעית ולא בטעם אישי או בהשוואת פיקסלים.</p>
<h2 id="להעמקה-עקומות-ליסאזו">להעמקה: עקומות ליסאז׳ו</h2>
<p dir="rtl" style="text-align:right">בחקירה זו, שהיא רשות ואינה נבחנת, נשתמש במתכון שרטוט מסופק עבור <span dir="ltr" class="math display">\[x(t)=A\cos(at+\delta),\qquad y(t)=B\sin(bt).\]</span> הזמן <span dir="ltr" class="math inline">\(t\)</span> בשניות, התדירויות הזוויתיות <span dir="ltr" class="math inline">\(a,b\)</span> ברדיאנים לשנייה, הפאזה <span dir="ltr" class="math inline">\(\delta\)</span> ברדיאנים, ו-<span dir="ltr" class="math inline">\(A,B\)</span> הן משרעות הקואורדינטות. העקומה מציגה שתי תנועות סינוסיות בלתי תלויות יחד; אין זה מסלול תחת כוח כבידה מרכזי.</p>
<p dir="rtl" style="text-align:right">נקבע משרעות ונשווה יחסי תדירות <span dir="ltr" class="math inline">\(1:1\)</span>, <span dir="ltr" class="math inline">\(2:1\)</span> ו-<span dir="ltr" class="math inline">\(3:2\)</span>. עבור יחס אחד נשנה את הפאזה מ-<span dir="ltr" class="math inline">\(0\)</span> ל-<span dir="ltr" class="math inline">\(\pi/2\)</span>. נשמור שני גרפים מנוגדים וננסח השערה על סגירות ועל סימטריה. נבדוק אם הארכת תחום הזמן משנה עקומה שנראתה פתוחה: חלון שרטוט סופי אינו הוכחה לאי-סגירות. הפעילות אינה מוסיפה תחביר חובה והיא נפרדת מהבונוס המזכה בנקודות.</p>
</div>
